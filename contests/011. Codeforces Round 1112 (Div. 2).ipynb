## A. Threshold Movement (00:10)

`-` $w_1$은 무조건 오른쪽으로 가야 한다. 그게 아니라면 가만히 있으면 안 되니 왼쪽으로 가는 것뿐이다

`-` 근데 $w_1$이 왼쪽으로 가게 되면 $1$번 위치를 채우기 위해 $w_2$도 왼쪽으로 가야 한다. 이를 재귀적으로 반복하면 다른 모든 원소가 왼쪽으로 가야만 한다. 하지만 결국 $w_n$도 왼쪽으로 움직여야 하니 $n$번 위치에 원소가 존재하지 않게 되어 $k$는 *perfect* 하지 않다

`-` 따라서 $w_1$은 오른쪽으로 가야 하고 $w_2$는 $1$번 위치를 채우기 위해 왼쪽으로 가야 한다. $1,2$번 위치가 채워졌으니 문제는 $w_3$부터 $w_n$까지 적절히 움직여 $3$번부터 $n$번 위치까지 채우는 걸로 바뀌었다

`-` 이는 $w_i$부터 $w_n$까지 적절히 움직여 $i$번부터 $n$번 위치까지 채우는 일반화된 문제를 푸는 것과 같다. 그리고 앞서 봤듯이 $w_i$는 오른쪽으로 $w_{i+1}$은 왼쪽으로 움직여야 한다. 이를 재귀적으로 수행하면 다음의 결론을 얻을 수 있다

`-` 짝을 맞추기 위해 $n$은 짝수여야 하고 각 원소의 움직임에 맞게 $k$의 최솟값과 최댓값을 추적한 뒤 대소 관계의 모순이 없다면 YES, 아니라면 NO를 출력하자

```python
def solve_testcase(array):
    n = len(array)
    if n % 2 == 1:
        return "NO"
    min_k, max_k = 0, 10**9 + 1
    for i, a in enumerate(array):
        if i % 2 == 0:
            max_k = min(a - 1, max_k)
        else:
            min_k = max(a + 1, min_k)
    if min_k <= max_k:
        return "YES"
    return "NO"


def solution():
    t = int(input())
    for _ in range(t):
        n = int(input())
        w = list(map(int, input().split()))
        answer = solve_testcase(w)
        print(answer)


solution()

# input
# 1
# 4
# 9 1 7 2
```

## B. Nikita and Books (00:28)

`-` $0$의 개수와 $1$의 개수 차이가 최대 $1$이도록 $s$를 구성해야 한다

`-` 만약 $k=n-1$이면 모든 원소가 동일해야 하니 조건을 만족하는 $s$는 존재하지 않는다. 따라서 $k < n-1$이라 생각하고 $n$과 $k$의 홀짝성에 따라 $s$를 구성해보자

`-` 우선 대칭성에 의해 $s$가 $0$으로 시작하든 $1$로 시작하든 상관없으므로 $0$으로 시작하는 걸로 통일하자

`-` 먼저 $n$이 짝수라고 해보자. $k=0$이라면 $s$는 단순히 $0$과 $1$이 번갈아 등장하는 형태이다

`-` $k$가 짝수이고 $m = \frac{k}{2} + 1$이라 하자. $0$이 연속해서 $m$번 등장하고 $1$이 연속해서 $m$번 등장한 뒤 남은 개수만큼 $0$과 $1$이 번갈아가며 등장하는 형태로 $s$를 구성할 수 있다

`-` $k$가 홀수라면 정확히 절반으로 나눌 수 없다. $s$는 $0$부터 시작하므로 하나 남는 걸 $1$에 투자하겠다. $m = \frac{k}{2} + 1$이라 하자. $0$이 연속해서 $m$번 등장하고 $1$이 연속해서 $m+1$번 등장한 뒤 남은 개수만큼 $0$과 $1$이 번갈아가며 등장하는 형태로 $s$를 구성할 수 있다

`-` 이제 $n$이 홀수라고 해보자. $k=0$이라면 $s$는 단순히 $0$과 $1$이 번갈아 등장하는 형태이다. 단, $0$으로 끝나서 $0$이 $1$보다 $1$개 더 많다

`-` $k$가 짝수이고 $m = \frac{k}{2} + 1$이라 하자. $0$이 연속해서 $m$번 등장하고 $1$이 연속해서 $m$번 등장한 뒤 남은 개수만큼 $0$과 $1$이 번갈아가며 등장하는 형태로 $s$를 구성할 수 있다. $n$과 $k$가 짝수인 경우와 다른 점이라곤 $s$가 $1$이 아닌 $0$으로 끝난다는 점이다

`-` $n,k$가 홀수인 경우와 $n$이 짝수이고 $k$가 홀수인 경우의 차이점은 $n,k$가 홀수라면 $s$가 $0$이 아닌 $1$로 끝난다는 점이다

`-` 이 외의 경우는 없으므로 정리한 케이스에 맞춰 $s$를 출력하면 된다

```python
def solve_testcase(n, k):
    if k == n - 1:
        return -1
    m = k // 2 + 1
    r = (n - k - 2) // 2
    if n % 2 == 0:
        if k % 2 == 0:
            return "0" * m + "1" * m + "01" * r
        return "0" * m + "1" * (m + 1) + "01" * r + "0"
    if k % 2 == 0:
        return "0" * m + "1" * m + "01" * r + "0"
    return "0" * m + "1" * (m + 1) + "01" * r


def solution():
    t = int(input())
    for _ in range(t):
        n, k = map(int, input().split())
        answer = solve_testcase(n, k)
        print(answer)


solution()

# input
# 4
# 6 2
# 6 3
# 5 2
# 7 1
```

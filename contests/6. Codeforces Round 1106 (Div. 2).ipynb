## A. Another Puzzle from Papyrus (00:07)

`-` 문제 잘못 읽어서 푸는데 조금 걸렸다. $a_i - 1$인데 $a_{i-1}$로 봤음;;;

`-` $a$를 재정렬 하는 건 $a$랑 $b$ 둘 다 재정렬 하는 것과 같다 (어차피 인덱스만 맞추면 상관 없으므로)

`-` $a$의 원소는 작아질 수만 있다. 따라서 $a$랑 $b$를 오름차순 정렬한 뒤 $a_i < b_i$라면 $a$를 $b$로 만들 수 없다

`-` 그렇지 않다면 시간은 $a_i - b_i$만큼 필요하다

`-` 정렬을 한 번 수행하는 경우와 그렇지 않은 경우 중 최소 시간이 정답이 된다

```python
def compute_execution_time(a, b, inf):
    return sum(a_i - b_i if a_i > b_i else inf for a_i, b_i in zip(a, b))


def solve_testcase(a, b, c, inf):
    t1 = compute_execution_time(a, b, inf)
    a.sort()
    b.sort()
    t2 = c + compute_execution_time(a, b, inf)
    t = min(t1, t2)
    if t == inf:
        t = -1
    return t


def solution():
    inf = float("inf")
    t = int(input())
    for _ in range(t):
        n, c = map(int, input().split())
        a = list(map(int, input().split()))
        b = list(map(int, input().split()))
        answer = solve_testcase(a, b, c, inf)
        print(answer)


solution()

# input
# 1
# 1 1
# 1
# 1
```

## B. Crimson Triples (Upsolving)

`-` 에디토리얼 봤다

`-` 첫 접근이 맞았는데 바보라 이상하게 생각해서 못풀었다. 정수론 문제는 최대한 일반화한 뒤 구체화하는 식으로 접근하자

`-` $\operatorname{lcm}(a,b)$와 $\operatorname{lcm}(b,c)$는 $b$의 배수이다. 따라서 $\gcd(\operatorname{lcm}(a,b),\operatorname{lcm}(b,c))$는 $b$의 배수이다. 그러므로 문제의 등호가 성립하려면 $\gcd(a,c)$가 $b$의 배수여야 한다. 이는 곧 $a$와 $c$가 $b$의 배수임을 의미한다

`-` $a$와 $c$는 같아도 되므로 고정된 $b$에 대해 가능한 $(a,c)$ 쌍의 수는 ${\left\lfloor\frac{n}{b}\right\rfloor}^2$이다 (여기까지 접근했는데 못풀었다면 믿으시겠습니까?)

`-` 가능한 $b$는 $1$부터 $n$까지이다. 따라서 정답은 $\sum\limits_{b=1}^{n}{\left\lfloor\frac{n}{b}\right\rfloor}^2$이다

`-` 으아아아아아아ㅏㅏㅏㅏㅏㅏㅏㅏㅏㅏㅏ

```python
def solve_testcase(n):
    return sum((n // b)**2 for b in range(1, n + 1))


def solution():
    t = int(input())
    for _ in range(t):
        n = int(input())
        answer = solve_testcase(n)
        print(answer)


solution()

# input
# 1
# 10
```

## C. Village Guilds (Upsolving)

`-` 가능한 모든 **guild**를 구하면 된다. 어떤 노드 $u$에 대해 $u$의 자식이 하나 뿐이라고 해보자

`-` 그럼 $u$에 의해 생성되는 **guild**는 $u$ 자신과 $u$의 자식에 의해 생성되는 **guild**이다

`-` 여기서 새로운 것은 자기 자신 뿐이다 (나머지는 서브 트리에서 이미 계산됐다)

`-` 만약 $u$의 자식이 둘 이상이라면 얘기는 달라진다

`-` $sd(u)$를 $u$의 서브 트리에 속한 노드 중 가장 깊은 노드의 깊이라고 하자

`-` 자식이 최소 $2$개는 있어야 병합을 해서 새로운 **guild**를 만들 수 있다. 그리고 $u$로부터 $h$만큼 떨어진 서브 트리의 노드는 변하지 않는다 (결정론적)

`-` 즉, 자식이 $2$개 이상이라면 새롭게 형성되는 **guild**는 자식의 수가 아닌 $sd$ 함수에만 영향을 받는다

`-` $u$의 자식들 $v_1,v_2,\dots, v_n$에 대해 $sd(v_1), sd(v_2), \dots, sd(v_n)$ 중 $2$번째로 큰 것을 $sd_2(u)$라 하자 (최댓값은 $sd_1(u)$)

`-` $h$는 $0$부터 $sd_2(u) - \operatorname{depth}[u]$까지 가능하다

`-` 따라서 $sd_2(u) - \operatorname{depth}[u] + 1$이 $u$에 의해 새롭게 생성되는 **guild**의 수이다

`-` 동적 계획법을 이용하여 모든 노드 $u$에 대해 $sd_2(u) - \operatorname{depth}[u] + 1$를 계산한 뒤 합산하면 정답이다

`-` 단, TLE, MLE를 방지하기 위해 비재귀 바텀-업 구현을 수행해야 한다. 가장 깊은 노드부터 한 단계씩 거슬러 올라가며 갱신하면 된다

```python
def count_guild(parents, depths, degrees, depth2nodes, max_depth):
    sd1 = depths[:]
    sd2 = depths[:]
    dp = [0] * len(parents)
    temp_sd1 = depths[:]
    temp_sd2 = depths[:]
    for depth in range(max_depth, -1, -1):
        for node in depth2nodes[depth]:
            sd1[node] = temp_sd1[node]
            sd2[node] = temp_sd2[node]
            dp[node] = sd2[node] - depths[node] + 1 if degrees[node] >= 2 else 1
            parent = parents[node]
            if temp_sd1[parent] < sd1[node]:
                temp_sd2[parent] = temp_sd1[parent]
                temp_sd1[parent] = sd1[node]
            elif temp_sd2[parent] < sd1[node]:
                temp_sd2[parent] = sd1[node]
    return sum(dp)


def dfs_depth(tree, parents, root):
    depths = [0] * len(tree)
    stack = [root]
    while stack:
        node = stack.pop()
        parent = parents[node]
        for child in tree[node]:
            if child == parent:
                continue
            stack.append(child)
            depths[child] = depths[node] + 1
    return depths


def solve_testcase(tree, parents, root):
    n = len(tree)
    depths = dfs_depth(tree, parents, root)
    max_depth = max(depths)
    depth2nodes = [[] for _ in range(max_depth + 1)]
    for node in range(root, n):
        depth2nodes[depths[node]].append(node)
    degrees = [0] * n
    for node in range(root + 1, n):
        degrees[parents[node]] += 1
    count = count_guild(parents, depths, degrees, depth2nodes, max_depth)
    return count


def solution():
    none, root = 0, 1
    t = int(input())
    for _ in range(t):
        n = int(input())
        p = [none] * 2
        p.extend(map(int, input().split()))
        tree = [[] for _ in range(n + 1)]
        for node, parent in enumerate(p):
            tree[node].append(parent)
            tree[parent].append(node)
        answer = solve_testcase(tree, p, root)
        print(answer)


solution()

# input
# 1
# 5
# 1 2 3 4
```

## D. Storming Arasaka (Upsolving)

`-` $1$을 제외한 $n$의 약수들을 조건에 맞춰 레이어에 분배할 때 사용하는 레이어의 수를 최소로 하면 된다

`-` $1$을 제외한 $n$의 약수를 작은 순으로 $d_1,d_2, \dots, d_m$이라 하자

`-` 레이어의 수는 최대 $m$이다. 큰 것부터 어떤 레이어에 포함시킬지 정하자 ($m$번 레이어부터 $1$번 레이어 순으로 고려)

`-` $d_i$를 레이어에 넣을 차례라고 하자. 만약 레이어가 비어있다면 넣으면 된다

`-` 그렇지 않다면 두 개의 조건을 만족해야 한다. 레이어의 들어있는 가장 큰 수를 $x$라고 하자

`-` $x$는 $d_i$보다 먼저 삽입됐으므로 더 크다. 만약 $g = \gcd(d_i, x)$에 대해 $g=x$라면 $d_i$는 $x$의 약수이고 $1$번 조건에 의해 같은 그룹에 포함될 수 없다

`-` $g=1$이어도 같은 그룹에 포함될 수 없다. 이 경우 다음 레이어로 넘어가면 되며 $g \ne 1$이고 $g \ne d_i$라면 해당 레이어에 삽입하면 된다

`-` $x$만 확인해도 상관 없는 이유는 다음과 같다. $x$랑 인접한 레이어 내 원소를 $y$라 하자

`-` $\gcd(x, d_i)= 1$이지만 $q=\gcd(y, d_i)$라 할 때 $1 < q < d_i$라고 해보자

`-` $x=ag, y = bgq, d_i=cq$의 꼴일 것이다 (단, $\gcd(x,y)=g$). 이 경우 $x, y, d_i$ 순으로 배치할 수 있으며 조건에 모순되지 않는다

`-` $\gcd(y, d_i) = q$이므로 $\gcd(bg, c)=1$이다. 그런데 추후 $bq$도 레이어에 포함시켜야 되는데 $bq$는 $y$의 약수이므로 $y$랑 같은 레이어가 아닌 더 작은 번호의 레이어에 넣어야 한다

`-` 이때 $\gcd(bq, d_i) = q$로 $1 < q < d_i$이다. 즉, $d_i$랑 $bq$는 같은 레이어에 포함될 수 있으며 사용된 레이어의 수는 같고 $d_i$의 위치만 바뀐 것이니 정답에 영향을 끼치지 않는다

`-` 만약 $d_i$의 약수들이 $bq$랑 같은 레이어에 있었다면? $d_i$의 약수들을 더 작은 번호의 레이어로 옮기면 된다

`-` 근데 이를 반복하다 보면 애초에 만들지 않아도 될 레이어를 추가로 만들어서 옮겨야 되는 게 아닐까?

`-` 그렇지 않다. 이를 반복하다 보면 결국엔 소수인 약수들을 고려하게 되는데 이들은 애초에 소수라 자기 혼자만 레이어에 포함되어야 한다 (그리고 $1$과 자기 자신을 제외하면 더 이상 고려할 약수도 없다)

`-` 따라서 $d_i$를 레이어에 포함시킬지 말지는 해당 레이어의 가장 큰 원소만 보고 판단해도 된다

`-` 근데 $d_i$가 $x$의 약수는 아니지만 $y$의 약수가 되어서 문제가 생길 수도 있지 않을까?

`-` 그러면 그냥 $d_i$를 다른 그룹으로 옮기면 되는데 위에서 확인했듯이 추가 레이어를 사용하지 않으므로 정답에 영향을 주지 않는다

`-` 모든 약수들을 레이어에 넣은 뒤 원소를 하나 이상 포함한 레이어의 수를 세면 이것이 정답이 된다

`-` 전체 알고리즘의 시간 복잡도는 최악의 경우 $O\left(t\left(\sqrt{n}\log \sqrt{n} + m^2\log n\right)\right)$이다

`-` 상수가 작고 약수 개수가 그리 많지 않아서 빠른 시간에 동작한다

```python
import math


def find_divisors(n):
    divisors = []
    for x in range(1, n + 1):
        if x * x > n:
            break
        if n % x != 0:
            continue
        divisors.append(x)
        q = n // x
        if x != q:
            divisors.append(q)
    return divisors


def solve_testcase(n):
    divisors = find_divisors(n)
    divisors.sort(reverse=True)
    divisors = divisors[:-1]
    m = len(divisors)
    layers = [[] for _ in range(m + 1)]
    for d in divisors:
        for layer in layers:
            if not layer:
                layer.append(d)
                break
            x = layer[0]
            g = math.gcd(x, d)
            if g == d or g == 1:
                continue
            layer.append(d)
            break
    count = sum(1 for l in layers if l)
    return count          


def solution():
    t = int(input())
    for _ in range(t):
        n = int(input())
        answer = solve_testcase(n)
        print(answer)


solution()

# input
# 1
# 10
```

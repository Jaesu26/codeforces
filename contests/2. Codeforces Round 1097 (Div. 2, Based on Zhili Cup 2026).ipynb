## A. Zhily and Array Operating (00:06)

`-` $i=n-1$부터 $i=1$까지 $a_{i+1}$이 음수면 연산을 패스하고 양수면 연산을 적용하자

```python
def solve_testcase(array, n):
    for i in range(n - 2, -1, -1):
        a_i = array[i]
        if array[i + 1] > 0:
            array[i] += array[i + 1]
    return sum(a > 0 for a in array)


def solution():
    t = int(input())
    for _ in range(t):
        n = int(input())
        a = list(map(int, input().split()))
        answer = solve_testcase(a, n)
        print(answer)


solution()

# input
# 1
# 3
# -2 -1 3
```

## B. Zhily and Mex and Max (00:36)

`-` $F_{\operatorname{mex}}=\sum\limits_{i=1}^{n} \operatorname{mex}(a_1,a_2,\dots, a_i)$, $F_{\max}=\sum\limits_{i=1}^{n} \max(a_1,a_2,\dots, a_i)$일 때 $F_{\operatorname{mex}} + F_{\max}$의 최댓값을 구해야 한다

`-` $F_{\operatorname{mex}}$를 최대화 하려면 $a$의 유니크한 원소들을 오름차순으로 배치한 뒤 남은 원소를 뒤에 배치하면 된다

`-` $F_{\max}$를 최대화 하려면 가장 큰 원소를 $a_1$으로 설정하면 된다. 나머지 원소들의 배치는 영향을 주지 않는다. $F_{\operatorname{mex}}$도 크게 하기 위해 나머지 원소들 중 유니크한 원소들을 고려하자. 만약 $a_1$을 포함한다면 유니크한 원소에서 $a_1$은 제외하자. 이제 유니크한 원소들을 오름차순으로 배치한 뒤 나머지를 뒤에 배치하면 된다

`-` 두 경우 중 최댓값이 $F_{\operatorname{mex}} + F_{\max}$의 최댓값이다

```python
def maximize_mex(array):
    array.sort()
    sum_ = f_max(array) + f_mex(array)[0]
    return sum_


def maximize_max(array):
    n = len(array)
    array.sort(reverse=True)
    a_max = array[0]
    unique = sorted(set(array), reverse=True)[1:]
    unique.sort()
    num_unique = len(unique)
    sum_, mex = f_mex([a_max, *unique])
    sum_ += n * a_max
    sum_ += mex * (n - num_unique - 1)
    return sum_


def f_max(array):
    sum_ = 0
    max_ = array[0]
    for a in array:
        if a > max_:
            max_ = a
        sum_ += max_
    return sum_


def f_mex(array):
    sum_ = 0
    mex = 0
    seen = {}
    for a in array:
        seen[a] = True
        if a == mex:
            mex += 1
        while seen.get(mex, False):
            mex += 1
        sum_ += mex
    return sum_, mex


def solve_testcase(array):
    result1 = maximize_mex(array)
    result2 = maximize_max(array)
    return max(result1, result2)


def solution():
    t = int(input())
    for _ in range(t):
        n = int(input())
        a = list(map(int, input().split()))
        answer = solve_testcase(a)
        print(answer)


solution()

# input
# 1
# 1
# 3
```

## C. Zhily and Bracket Swapping (00:54)

`-` $a$와 $b$ 각각에 대해 `(` 개수와 `)` 개수의 차이를 기록하자. 이것이 음수가 되는 순간 올바른 괄호 문자열이 될 수 없다

`-` $a_1, b_1$부터 시작해서 `(`와 `)`를 잘 나누어 가지면 된다

```python
def solve_testcase(a, b):
    a_gap = 0  # count: '(' - ')'
    b_gap = 0
    for a_i, b_i in zip(a, b):
        if a_i == b_i and a_i == "(":
            a_gap += 1
            b_gap += 1
            continue
        if a_i == b_i and a_i == ")":
            a_gap -= 1
            b_gap -= 1
            if a_gap >= 0 and b_gap >= 0:
                continue
            return "NO"
        if a_gap > b_gap:
            a_gap -= 1
            b_gap += 1
            continue
        a_gap += 1
        b_gap -= 1
        if a_gap >= 0 and b_gap >= 0:
            continue
        return "NO"
    if a_gap == 0 and b_gap == 0:
        return "YES"
    return "NO"


def solution():
    t = int(input())
    for _ in range(t):
        n = int(input())
        a = input().rstrip()
        b = input().rstrip()
        answer = solve_testcase(a, b)
        print(answer)


solution()

# input
# 1
# 2
# ()
# ()
```

## D. Zhily and Barknights

`-` 에디토리얼 봤다. 근데 본 지 2달 됐다...

`-` 인버전의 개수를 세는 과정을 생각해보자. $b'$을 하나 정한 뒤 $c$를 생성한다

`-` 그 뒤 임의의 $i,j\,(i<j)$에 대해 $c_i > c_j$라면 인버전 개수를 하나 증가시킨다

`-` 모든 $i,j$ 쌍을 순회한 뒤의 인버전 개수가 고정된 $b'$에 의해 생성된 $c$의 인버전 개수이다

`-` 이를 모든 $b'$에 대해 반복하여 인버전 개수를 합산한 뒤 $n!$로 나누면 된다

`-` 이때 가능한 $b'$의 수는 $n!$이므로 브루트 포스로는 해결할 수 없다

`-` 위의 과정을 다른 측면에서 생각하면 다음과 같다

`-` 고정된 $(i,j)$에 대해 가능한 모든 $b'$을 떠올리자 (단, $i<j)$. 각 $b'$마다 $a_i \cdot b'_i > a_j \cdot b'_j$를 만족하면 인버전 개수를 하나 증가시키자

`-` 모든 $b'$에 대해 계산을 마친 결과가 고정된 $(i,j)$에 의해 생성된 $c$의 인버전 개수이다

`-` 이를 모든 $(i,j)$에 대해 반복하여 인버전 개수를 합산한 뒤 $n!$으로 나누면 된다

`-` 그런데 고정된 $(i,j)$ 입장에서 가능한 $b'_i, b'_j$의 수는 $n(n-1)$이다. $a_i \cdot b'_i > a_j \cdot b'_j$를 만족하면 인버전인 것인데 이 부등식은 $i,j,b'_i,b'_j$에만 영향을 받는다

`-` $n$개의 원소 중 $i,j$번째 원소만 고정됐으므로 이를 만족하는 배열의 개수는 $(n-2)!$이다

`-` 따라서 $a_i \cdot b'_i > a_j \cdot b'_j$를 만족하면 $(i,j)$에 의해 생성된 $c$의 인버전 개수는 $(n-2)$!이다 (아니라면 $0$)

`-` 나이브하게 계산하면 $O\left(n^4\right)$에 가능한 모든 $c$의 인버전 개수를 계산할 수 있다

`-` $n$이 최대 $2000$이므로 시간 복잡도를 더 줄여보자. 인버전일 조건은 $a_i \cdot b'_i > a_j \cdot b'_j$를 만족하는 것이다

`-` 이때 $a_i, a_j$는 $i,j$가 정해지면 고정이므로 $\frac{a_i}{a_j } > \frac{b'_j}{b'_i}$와 같이 나타냈을 때 가능한 모든 $b'_i, b'_j$를 계산한 뒤 $\frac{b'_j}{b'_i}$를 기준으로 오름차순 정렬해놓으면 $\frac{a_i}{a_j } > \frac{b'_j}{b'_i}$를 만족하는 $(b'_i, b'_j)$의 수를 이분 탐색으로 $O\left(\log n^2\right) = O(\log n)$에 계산할 수 있다

`-` 따라서 가능한 모든 $c$의 인버전 개수를 $O\left(n^2\log n\right)$에 얻을 수 있다

`-` 마무리로 적절한 모듈러 연산만 취해주면 된다

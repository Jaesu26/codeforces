## A. The Best Card (00:03)

`-` 다른 모든 카드와 대결했을 때 이기는 카드의 값을 $z$라 하자. $z$는 약수가 존재하면 안 되니 소수여야 한다

`-` $z$가 소수이니 $z$보다 큰 값을 가지는 카드가 존재하면 패배한다. 따라서 $z$가 가장 커야하므로 $z = n + 1$이다

`-` 즉, $n+1$이 소수이면 YES, 아니면 NO이다

```python
def solve_testcase(n):
    x = n + 1
    for i in range(2, x + 1):
        if i * i > x:
            break
        if x % i == 0:
            return "NO"
    return "YES"


def solution():
    t = int(input())
    for _ in range(t):
        n = int(input())
        answer = solve_testcase(n)
        print(answer)


solution()

# input
# 2
# 3
# 4
```

## B. Hypercarp and the Control Panel (00:30)

`-` 배열 $a$를 연속된 동일한 원소 그룹으로 나누자. 인접한 원소가 서로 달라야 하니 기본적으로 그룹의 개수를 정답으로 취할 것이다. 즉, 그룹마다 하나의 원소만 남기고 나머지는 삭제하는 것이다

`-` 이제 최대 $1$번의 교환을 통해 추가적으로 삭제하지 않고 남길 수 있는 원소의 개수를 구하자. 기본적으로 크기가 $2$ 이상인 그룹을 고려하자. 만약 $i$번 그룹의 오른쪽에 있는 $i+1$번 그룹의 크기가 $2$ 이상이라면 교환을 통해 $2$개의 원소를 덜 삭제할 수 있다

`-` 그렇지 않다면 $i-1, i+1$번 그룹을 확인하자. 이들의 크기가 $1$이고 그룹의 처음 또는 끝에 위치한다면 교환을 통해 $1$개의 원소를 덜 삭제할 수 있다. 처음 또는 끝이 아니라면 각각에 대해 $i-2, i + 2$번 그룹을 확인하자. 이 그룹의 원소가 $i$번 그룹의 원소와 다르다면 교환을 통해 $1$개의 원소를 덜 삭제할 수 있다

`-` 고려할 경우가 많으니 구현할 때 주의하자

```python
from itertools import groupby


def solve_testcase(array):
    n = len(array)
    groups = [list(group) for _, group in groupby(array)]
    m = len(groups)
    if m == 1:
        return 1
    n_additional = 0
    for i in range(m):
        if len(groups[i]) < 2:
            continue
        if i + 1 < m and len(groups[i + 1]) >= 2:
            n_additional = 2
            break
        elif i - 1 == 0 or i + 1 == m - 1:
            n_additional = 1
        elif i - 2 >= 0 and groups[i][0] != groups[i - 2][0]:
            n_additional = 1
        elif i + 2 < m and groups[i][0] != groups[i + 2][0]:
            n_additional = 1
    return m + n_additional


def solution():
    t = int(input())
    for _ in range(t):
        n = int(input())
        a = list(map(int, input().split()))
        answer = solve_testcase(a)
        print(answer)


solution()

# input
# 1
# 5
# 2 2 3 3 1
```

`-` 배열을 임의의 기준에 따라 연속된 동일한 원소 그룹으로 나눠야 할 땐 `itertools.groupby`를 적극 사용하자

## C. Sum of Distinct Values in a Matrix (01:13)

`-` $a$와 $b$를 포함해 최대 $n + m - 1$개의 유니크한 원소를 고를 수 있다. 이때 $a$에서 $n$개 고르고 $b$에서 $m-1$개 고르거나 $a$에서 $n-1$개 고르고 $b$에서 $m$개를 고르면 된다 (버츄얼 땡큐!)

`-` 매트릭스 비용을 최대화하려면 큰 원소부터 사용해야 한다. $a$와 $b$의 교집합이 공집합이라면 원소 선택이 독립적이지만 그렇지 않다면 고려할 게 생긴다. 어떤 원소 $k$를 선택해야 하는데 이를 $a$에서 고를 것인가 $b$에서 고를 것인가?

`-` 이를 고려하기 위해 $a$와 $b$를 하나의 배열로 병합한 뒤 중복 원소는 $1$개만 남기자. 정렬하여 큰 원소부터 고려할 것이다. 만약 공통으로 존재하는 원소라면 기본적으로 남은 선택 횟수가 큰 쪽이 부담하자. 그리고 각 배열이 부담한 횟수를 기록해두자

`-` 둘 중 하나에만 존재하는 원소라면 해당 배열에서 선택하면 된다. 만약 남은 선택 횟수가 $0$이라면 여태까지 해당 배열이 부담한 횟수를 보자. 부담한 횟수가 $0$이라면 해당 원소를 고를 수 없지만 아니라면 대신 상대 배열에게 부담시키자 (단, 상대 배열의 남은 선택 횟수가 $1$ 이상이여야 한다). 이유는 다음과 같다

`-` 현재 고려하고 있는 원소는 앞으로 고려할 원소 중 가장 큰 원소이다. 따라서 앞으로 상대 배열에서 고를 원소의 개수를 $1$ 줄이더라도 현재 고려 중인 원소를 선택하는 게 귀류법에 의해 이득이다

`-` 두 배열의 남은 선택 횟수가 $0$이거나 모든 배열의 원소를 고려했다면 탐색을 끝내면 된다. 여태까지 고른 원소의 합이 최대 매트릭스 비용이 된다

```python
def maximize_matrix_cost(a_count, b_count, array, a_set, b_set):
    a_loss = b_loss = 0
    cost = 0
    for k in array:
        if a_count + b_count <= 0:
            break
        is_in_a = str(k) in a_set
        is_in_b = str(k) in b_set
        if is_in_a and not is_in_b:  
            if a_count > 0: 
                cost += k
                a_count -= 1
            elif a_loss > 0 and b_count > 0:
                a_loss -= 1
                b_loss += 1
                cost += k
                b_count -= 1
            continue
        if not is_in_a and is_in_b: 
            if b_count > 0: 
                cost += k
                b_count -= 1
            elif b_loss > 0 and a_count > 0:
                a_loss += 1
                b_loss -= 1
                cost += k
                a_count -= 1
            continue
        if a_count > b_count: 
            cost += k
            a_loss += 1
            a_count -= 1
        else:
            cost += k
            b_loss += 1
            b_count -= 1
    return cost


def solve_testcase(n, m, a, b):
    a_set = set(map(str, a))
    b_set = set(map(str, b))
    array = a + b
    array = set(map(str, array))
    array = list(map(int, array))
    array.sort(reverse=True)
    cost1 = maximize_matrix_cost(n, m - 1, array, a_set, b_set)
    cost2 = maximize_matrix_cost(n - 1, m, array, a_set, b_set)
    return max(cost1, cost2)


def solution():
    t = int(input())
    for _ in range(t):
        n, m, x, y = map(int, input().split())
        a = list(map(int, input().split()))
        b = list(map(int, input().split()))
        answer = solve_testcase(n, m, a, b)
        print(answer)


solution()

# input
# 1
# 2 2 2 2
# 1 4
# 2 3
```

## D. Hypercarp and Interdimensional Jumps (01:57)

`-` $s$를 바탕으로 이동한 우주선이 최종적으로 $(p, q)$에 멈출 때 $(p - x)^2 + (q - y)^2$가 모든 이동 경로 집합 중 최소여야 한다

`-` 점프 벡터의 각 성분은 감소하지 않으며 누적됨에 유의하자. 현재를 포함해 $j$번의 이동만 남았다고 할 때 $a$를 $1$ 증가시키는 것은 첫 번째 좌표 방향으로 총 $j$만큼 점프하는 것과 같다

`-` 남은 이동 횟수에 따라 점프량이 변하니 이를 고정시키고자 최적 이동 횟수를 $k$라 가정하자. 그럼 $p + q = 1 + 2 + \dots + k - 1 + k = \frac{k(k + 1)}{2}$를 만족한다 (모든 점프 크기의 합)

`-` 유클리드 거리 차이를 최소화하기 위해 $\frac{k(k + 1)}{2} \le x + y$를 만족하는 $k$ 중 최댓값인 $k'$을 구하자 (나는 콘테스트 중엔 이를 브루트 포스로 구했는데 제미나이한테 효율적인 방법을 물어보니 이차부등식을 이용해 $O(1)$에 구하는 법을 알려줬다, 근데 브루트 포스도 $O(20000)$이라 TLE가 발생하진 않는다)

`-` $r = x + y - (p + q) = x + y - \frac{k'(k' + 1)}{2}$이라 하자. 나는 정확한 $(p,q)$를 모른다. 하지만 $(x - r, y), (x - r + 1, y - 1), \dots, (x - 1, y - r + 1), (x, y - r)$ 중에 하나인 것은 안다. 그리고 이 중 $(p-x)^2 + (q - y)^2$를 최소화하는 건 $p= x - \left\lfloor\frac{r}{2}\right\rfloor, q = y - r + \left\lfloor\frac{r}{2}\right\rfloor$일 때이다 ($p= x - r + \left\lfloor\frac{r}{2}\right\rfloor, q = y - \left\lfloor\frac{r}{2}\right\rfloor$도 마찬가지)

`-` 이제 최적의 $(p,q)$를 구했으니 이를 기반으로 $s$를 재구성하면 된다. 즉, $1,2,\dots, k' - 1, k'$중 적절히 몇 개를 골라 합이 $p(= x - \left\lfloor\frac{r}{2}\right\rfloor)$가 되도록 만들면 된다. 이건 항상 가능하며(생각해보면 당연하며 수학적 귀납법으로 증명 가능) $j$를 $k'$부터 $1$까지 순회하며 $p \ge j$라면 배열에 $X$를 추가한 뒤 $p$를 $p-j$로 갱신하고 그렇지 않다면 $Y$를 추가하자. 순회를 마친 뒤 배열을 결합하면 $s$가 된다

```python
import math


def solve_testcase(x, y):
    z = x + y
    n_moves = (math.isqrt(8 * z + 1) - 1) // 2
    total = (n_moves * (n_moves + 1)) // 2
    r = z - total
    r_x = x - (r // 2)
    track = []
    for j in range(n_moves, 0, -1):
        if r_x >= j:
            track.append("X")
            r_x -= j
        else:
            track.append("Y")
    s = "".join(track)
    return s


def solution():
    t = int(input())
    for _ in range(t):
        x, y = map(int, input().split())
        answer = solve_testcase(x, y)
        print(answer)


solution()

# input
# 1
# 4 2
```

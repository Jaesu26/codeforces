## A. Divide and Conquer (00:01)

`-` $x$가 $y$의 배수인지 판단하면 된다

```python
def solve_testcase(x, y):
    if x % y == 0:
        return "YES"
    return "NO"


def solution():
    t = int(input())
    for _ in range(t):
        x, y = map(int, input().split())
        answer = solve_testcase(x, y)
        print(answer)


solution()

# input
# 1
# 2 1
```

## B. Good times Good times (00:13)

`-` 이거 원래 6분 컷 냈는데 다 풀어놓고 개멍청 실수해서 1번 틀렸다

`-` $x$의 자릿수를 $d$라 하자

`-` $y=10^d + 1$로 설정하면 $y$는 $0$과 $1$만 쓰이므로 good이고 $xy = x + x\cdot 10^d$으로 $x$를 $2$번 이어붙인 격이라 good이다

`-` 처음에 $y=10$ 같은 걸 생각했고 이를 발전시켜 $xy$를 $x$를 $2$번 이어붙인 꼴로 만들 수 있지 않을까 생각하다 위 형태를 찾았다

`-` $x$와 $y$의 범위도 힌트가 됐다

```python
def solve_testcase(x):
    d = len(str(x))
    y = 10**d + 1
    return y


def solution():
    t = int(input())
    for _ in range(t):
        x = int(input())
        answer = solve_testcase(x)
        print(answer)


solution()

# input
# 1
# 6767
```

## C. RemovevomeR (01:11)

`-` 어떻게 풀지 바로 생각이 안 났다

`-` 우선 $s$가 $0$ 또는 $1$로만 되어있으면 정답은 $1$이다

`-` 그리고 이게 보니까 부분 문자열이 $00, 11$꼴이면 팰린드롬이니 길이를 $1$ 줄일 수 있고 $01, 01$꼴이면 좌우에 $0$ 또는 $1$이 있을텐데 그럼 길이가 $2$ 이상인 팰린드롬이 존재하므로 역시 길이를 $1$ 줄일 수 있다

`-` 이를 반복하다 보면 결국엔 $01$ 또는 $10$만 남는다. 그래서 정답의 상한은 $2$이다

`-` 여기서 좀 더 발전시켜서 $s$의 길이가 짧을 때 정답이 어떻게 되는지 계산해봤다

`-` 정확한 증명은 모르겠지만 $s$의 형태가 $0\dots 01 \dots 1$ 또는 $1\dots 10 \dots 0$만 아니라면 최종 길이를 $1$로 만들 수 있을 것 같았고 제출하니 맞았다

`-` 추후 에디토리얼 한번 확인하자

```python
def solve_testcase(s):
    unique = set(s)
    if len(unique) == 1:
        return 1
    turn = 0
    prev = s[0]
    for s_i in s:
        if s_i == prev:
            continue
        turn += 1
        prev = s_i
    if turn > 1:
        return 1
    return 2


def solution():
    t = int(input())
    for _ in range(t):
        n = int(input())
        s = input().rstrip()
        answer = solve_testcase(s)
        print(answer)


solution()

# 1
# 4
# 1000
```

## D. An Alternative Way (00:27)

`-` 만약 $l=r$이라면 $i=l=r$이므로 $i$는 하나뿐이고 $i-l$은 짝수이다

`-` 따라서 $a_i$만 원하는 만큼 증가시킬 수 있다

`-` 문제는 $a_i$가 $b_i$보다 커서 $a_i$를 감소시켜야 할 때이다

`-` 그러기 위해선 $i-l$이 홀수여야 한다. 영향을 최소로 하기 위해 $l=i-1$로 놓으면 $a_i$는 감소하고 $a_{i-1}$은 증가하게 된다

`-` 처음부터 끝까지 순회하며 $a_i$를 감소시킬 수 있는 만큼 최대한 감소시키자. 그래야 $l=i, r=i+1$일 때 $a_{i+1}$을 최대한 감소시킬 수 있다

`-` 정확히는 $a_{i}$가 $b_{i}$를 초과하지 않을 때까지 $a_{i+1}$을 감소시킬 수 있다. 만약 이것이 불가능하다면 NO이고 가능하다면 YES이다

```python
def solve_testcase(a, b):
    stack = 0
    for a_i, b_i in zip(a, b):
        if a_i <= b_i:
            stack += b_i - a_i
            continue
        need = a_i - b_i
        if stack < need:
            return "NO"
        stack -= need
    return "YES"


def solution():
    t = int(input())
    for _ in range(t):
        n = int(input())
        a = list(map(int, input().split()))
        b = list(map(int, input().split()))
        answer = solve_testcase(a, b)
        print(answer)


solution()

# input
# 1
# 1
# 2
# 22
```

## E. Fair and Square (Upsolving)

`-` 에디토리얼 봤다. 침착했으면 풀만 했는데 아깝다 (실력이 부족해서 그럼)

`-` $x$를 루트 노드로 두었을 때 $x = \operatorname{lca}(u, v) = \operatorname{lca}(v, w) = \operatorname{lca}(w, u)$가 되도록 $u,v,w$를 설정하자

`-` 그럼 $f(u,v,w) = p(u, v) \cdot p(v, w) \cdot p(w, u)$에서 $a_x$만 $3$번 선택되고 나머지는 $2$번 선택된다 (당연하지만 경로 외에 존재하는 정점은 선택 안 됨)

`-` 즉, $f(u,v,w)$를 **perfect square**로 만들기 위해선 $a_x$가 제곱수여야 한다. 따라서 제곱수를 가지고 있는 노드만 $x$로 고려하면 된다 

`-` $x$의 자식 $c_1,c_2,\dots, c_m$에 대해 각 자식들의 서브 트리 크기를 $s_{c_1}, s_{c_2} \dots, s_{c_m}$라고 하자

`-` $x = \operatorname{lca}(u, v) = \operatorname{lca}(v, w) = \operatorname{lca}(w, u)$를 만족하기 위해선 임의의 자식의 서브 트리에서 둘 이상의 노드를 고르면 안 된다

`-` $u,v,w$ 중 $x$가 포함되거나 포함되지 않을 수 있으니 경우의 수는 다음과 같다

`-` 만약 셋 중 하나가 $x$라면 가능한 조합의 수는 $\sum\limits_{i<j}s_{c_i} s_{c_j}$이다

`-` 그렇지 않다면 가능한 조합의 수는 $\sum\limits_{i<j<k}s_{c_i} s_{c_j} s_{c_k}$이다

`-` 여기까진 도출했지만 이를 선형 시간에 풀 수 있는 방법을 찾지 못해 못풀고 에디토리얼을 확인했다

`-` 이는 간단한 동적 계획법으로 해결할 수 있다

`-` $i$번째 원소까지 고려했을 때 원소 $j$개 곱들의 합을 저장하자

`-` $i$번째 원소를 사용하거나 사용하지 않을 수 있므로 $\operatorname{dp}[i][j] = \operatorname{dp}[i - 1][j]+ \operatorname{dp}[i - 1][j - 1] \cdot s_{c_{i}}$이다 (단, $1\le j\le 3,\, \operatorname{dp}[i][0] = 1$)

`-` $x$의 차수를 $m$이라 할 때 동적 계획법의 시간 복잡도는 $O(m)$이다.  그리고 임의의 노드 $x$가 루트일 때 자식들의 서브 트리 크기는 $1$이 루트 노드라고 가정했을 때 계산한 $s_{c_1}, s_{c_2} \dots, s_{c_m}, n - s_x$이다

`-` 모든 노드의 차수의 합은 $2(n-1)$이므로 전체 알고리즘의 시간 복잡도는 $O(n)$이다

```python
def dfs_setup(tree, root):
    n = len(tree)
    parents = [0] * n
    dfs_order = []
    stack = [root]
    while stack:
        node = stack.pop()
        dfs_order.append(node)
        for child in tree[node]:
            if child == parents[node]:
                continue
            parents[child] = node
            stack.append(child)
    return parents, dfs_order


def compute_subtree_sizes(tree, parents, dfs_order):
    subtree_sizes = [1] * len(tree)
    for node in reversed(dfs_order):
        subtree_sizes[parents[node]] += subtree_sizes[node]
    return subtree_sizes


def count_cases(array):
    count = 0
    dp = [0] * 4
    dp[0] = 1
    for a in array:
        for j in range(3, 0, -1):
            dp[j] += dp[j - 1] * a
    count += dp[2] + dp[3]
    return count


def solve_testcase(tree, values, root):
    n = len(tree)
    parents, dfs_order = dfs_setup(tree, root)
    subtree_sizes = compute_subtree_sizes(tree, parents, dfs_order)
    total = 0
    for node, value in enumerate(values, start=1):
        sqrt = round(value**0.5)
        if sqrt * sqrt != value:
            continue
        elements = [subtree_sizes[child] for child in tree[node] if child != parents[node]]
        reminder = n - subtree_sizes[node] - 1
        if reminder > 0:
            elements.append(reminder)
        count = count_cases(elements)
        total += count
    return total


def solution():
    root = 1
    t = int(input())
    for _ in range(t):
        n = int(input())
        a = list(map(int, input().split()))
        tree = [[] for _ in range(n + 1)]
        for _ in range(n - 1):
            u, v = map(int, input().split())
            tree[u].append(v)
            tree[v].append(u)
        answer = solve_testcase(tree, a, root)
        print(answer)


solution()

# input
# 1
# 3
# 1 1 1
# 1 2
# 1 3
```

`-` 트리 DP를 비재귀로 구현할 때 DFS order 역순으로 갱신하면 효율적인 걸 제미나이가 알려줬다

`-` 기존 비재귀 구현보다 $25\%$ 빨라졌으니 앞으로도 유용하게 사용하면 될 것 같다

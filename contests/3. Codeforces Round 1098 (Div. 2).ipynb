## A. Marisa Steals Reimu's Takeout (00:05)

`-` 적당한 부분집합을 골라 합이 $3$의 배수가 되도록 할 때 가능한 부분집합의 크기를 작게 해야 이득이다

`-` 일단 $0$의 개수만큼 정답에 누적하자

`-` $1$은 $3$개 있어야 합이 $3$의 배수이며 $2$도 마찬가지이다

`-` 하지만 $1$과 $2$가 있으면 $2$개만으로 합이 $3$의 배수가 되니 따로 나누는 것보다 이득이다

`-` 그 후 남은 원소의 개수를 $3$으로 나눈 몫만큼 정답에 누적해주면 된다

```python
def solve_testcase():
    n = int(input())
    array = list(map(int, input().split()))
    answer = 0
    num_zeros = len([a for a in array if a == 0])
    num_ones = len([a for a in array if a == 1])
    num_twos = len([a for a in array if a == 2])
    answer += num_zeros
    if num_ones == num_twos:
        answer += num_ones
    elif num_ones > num_twos:
        answer += num_twos
        num_ones -= num_twos
        answer += num_ones // 3
    else:
        answer += num_ones
        num_twos -= num_ones
        answer += num_twos // 3
    print(answer)


def solution():
    t = int(input())
    for _ in range(t):
        solve_testcase()


solution()

# input
# 1
# 1
# 0
```

## B. Remilia Plays Soku (00:23)

`-` $n$이 $2$ 또는 $3$이면 정답은 $1$이다 (이것 때문에 1번 틀렸다)

`-` 그게 아니라면 일단 Reimu를 왼쪽 또는 오른쪽 중 최단 경로로 Remilia의 위치까지 이동시키자

`-` Remilia는 $k$번 이동할 수 있으니 Reimu는 Remilia를 잡기 위해 추가로 $k$번 더 이동해야 한다

`-` 따라서 $\min\left(|x_1 - x_2|, n - |x_1 - x_2|\right) + k$가 정답이 된다

```python
def solve_testcase():
    n, x1, x2, k = map(int, input().split())
    if n == 2 or n == 3:
        print(1)
        return
    diff = min(abs(x1 - x2), n - abs(x1 - x2))
    print(diff + k)


def solution():
    t = int(input())
    for _ in range(t):
        solve_testcase()


solution()

# input
# 1
# 2 1 2 11
```

## C1. Cirno and Number (Easy Version) (Upsolving)

`-` 왜 못풀었을까....

`-` $a$의 자릿수가 $\operatorname{digit}(a)$일 때 정답으로 가능한 $b$의 자릿수 범위는 $\max(\operatorname{digit}(a) - 1, 1) \sim \operatorname{digit}(a) + 1$이다

`-` 각 경우마다 $|a-b|$를 최소로 해나가면 된다

`-` 현재 $d$ 자리에 $d_1$을 사용할지 $d_2$를 사용할지 고민 중이라고 해보자. 만약 $d_1$을 사용한다면 이후로 $d_2$만 사용했을 때가 $b$를 최대로 할 수 있다

`-` 반대로 $d_2$를 사용한다면 앞으로 $d_1$만 사용했을 때 $b$가 최소이다. 두 경우를 비교해서 $|a-b|$가 더 작은 경우를 채택하자

`-` 이유는 다음과 같다

`-` 만약 $a$가 전자보다 작거나 후자보다 크다면 채택되지 않은 경우는 앞으로 $|a-b|$를 채택된 경우보다 더 작게 만들 수 없다

`-` 그렇지 않으면 $a$는 전자와 후자의 사이 값이다. $\text{전자} \le a \le \text{후자}$인 경우이니 전자는 앞으로 수를 최대한 키우고 후자는 최대한 줄여야 $a$와 가까워진다

`-` 따라서 전자를 최대한 키운 경우와 후자를 최대한 줄인 경우를 비교했을 때 $|a-b|$가 더 작은 경우를 채택하면 된다

`-` 예외 사항으로 만약 $d$가 일의 자리가 아닌 맨 앞자리라면 $0$이 올 수 없다. 이 경우 $d_1=0$이라면 무조건 $d_2$를 채택하자

`-` 이를 그대로 구현하면 정답이다

```python
def solve_testcase(a, digits):
    a_digit = len(str(a))
    d_1, d_2 = digits
    gap = INF
    for b_digit in range(max(a_digit - 1, 1), a_digit + 2):
        b = 0
        for ith_digit in range(b_digit, 0, -1):
            exp = 10**(ith_digit - 1)
            if ith_digit == b_digit and ith_digit != 1 and d_1 == 0:
                b += d_2 * exp
                continue
            b_min = b + d_1 * exp + (int(str(d_2) * (ith_digit - 1)) if ith_digit > 1 else 0)
            b_max = b + d_2 * exp + (int(str(d_1) * (ith_digit - 1)) if ith_digit > 1 else 0)
            if abs(a - b_min) < abs(a - b_max):
                b += d_1 * exp
            else:
                b += d_2 * exp
        gap = min(abs(a - b), gap)
    return gap


def solution():
    global INF
    INF = float("inf")
    t = int(input())
    for _ in range(t):
        a, n = map(int, input().split())
        digits = list(map(int, input().split()))
        answer = solve_testcase(a, digits)
        print(answer)


solution()

# input
# 1
# 1234 2
# 1 2
```

## C2. Cirno and Number (Hard Version) (Upsolving)

`-` Easy Version에선 $n=2$였지만 여기선 $1\le n \le 10$이다

`-` 하지만 해결 논리는 거의 비슷하다

`-` $a$의 자릿수가 $\operatorname{digit}(a)$일 때 정답으로 가능한 $b$의 자릿수 범위는 $\max(\operatorname{digit}(a) - 1, 1) \sim \operatorname{digit}(a) + 1$이다

`-` 현재 $d$ 자리에 어떤 숫자를 사용할지 고민 중이라고 해보자. 만약 해당 숫자를 선택했을 때 $a$보다 크다면 앞으로는 가장 작은 숫자를 선택해야 $a$와의 차이를 최소로 할 수 있다

`-` 그렇지 않고 $a$보다 작다면 앞으로는 가장 큰 숫자를 선택해야 한다. 만약 여태까지 고른 숫자와 앞에서부터 $a$의 $d+1$자리까지가 동일하다면 조금 복잡해진다. 이 경우 현재 선택에 따라 $a$보다 작아질 수도 있고 커질 수도 있고 같을 수도 있다. 따라서 가능한 모든 숫자를 순회하며 다음 자릿수로 넘어가서 작업을 재귀적으로 수행하자

`-` 모든 경우를 비교해서 $|a-b|$가 가장 작은 경우를 채택하면 된다

`-` 예외 사항으로 만약 $d$가 일의 자리가 아닌 맨 앞자리라면 $0$이 올 수 없다. 이 경우 $0$이 존재한다면 스킵하자

```python
def dfs(a, b, curr_digit, digits, is_less, is_greater, is_leading_zero):
    if curr_digit == 0:
        return b
    if is_less:
        return dfs(a, b * 10 + digits[-1], curr_digit - 1, digits, is_less, is_greater, is_leading_zero)
    if is_greater:
        return dfs(a, b * 10 + digits[0], curr_digit - 1, digits, is_less, is_greater, is_leading_zero)
    a_prefix = a // 10**(curr_digit - 1)
    gap = INF
    b_answer = None
    for d_i in digits:
        if is_leading_zero and d_i == 0:
            continue
        b_next = b * 10 + d_i
        b_complete = dfs(a, b_next, curr_digit - 1, digits, is_less or (b_next < a_prefix), is_greater or (b_next > a_prefix), False)
        if abs(a - b_complete) < gap:
            gap = abs(a - b_complete)
            b_answer = b_complete
    return b_answer    


def solve_testcase(a, digits):
    d_min, d_max = digits[0], digits[-1]
    if d_min == 0 and d_max == 0:
        return a
    a_digit = len(str(a))
    gap = INF
    for b_digit in range(max(a_digit - 1, 1), a_digit + 2):
        is_leading_zero = True if b_digit > 1 else False
        b = dfs(a, 0, b_digit, digits, False, False, is_leading_zero)
        if abs(a - b) < gap:
            gap = abs(a - b)
    return gap


def solution():
    global INF
    INF = float("inf")
    t = int(input())
    for _ in range(t):
        a, n = map(int, input().split())
        digits = list(map(int, input().split()))
        answer = solve_testcase(a, digits)
        print(answer)


solution()

# input
# 1
# 0 3
# 0 1 2
```

## A. Koshary (00:05)

`-` 첫 대회라 제출 미숙함 + 영어 해석 이상하게 함의 콜라보로 5분이나 걸렸다

`-` 실제 제출할 땐 빠른 입출력을 사용했다 (다른 것도 동일)

```python
def solve_testcase(x, y):
    if x % 2 == 1 and y % 2 == 1:
        return "NO"
    return "YES"


def solution():
    t = int(input())
    for _ in range(t):
        x, y = map(int, input().split())
        answer = solve_testcase(x, y)
        print(answer)


solution()

# input
# 1
# 2 1
```

## B. Party Monster (00:18)

`-` 도중에 Submit 버튼이 비활성화 되어서 나는 연속해서 제출할 수 없는 줄 알았다

`-` 그게 아니었고 Choose problem을 제대로 설정하지 않아서 그런 거였다. 이것 때문에 5분은 날렸다...

`-` 삭제한 괄호들을 원하는 위치에 배치할 수 있으니 `(`와 `)`의 짝이 맞는지만 판단하면 된다

`-` 입력 받을 때 공백을 제거 안 하는 바람에 1번 틀렸다. 다음부턴 rstrip으로 잘 제거해주자. 나는 else문 대신 elif문을 사용해서 통과했다

```python
def solve_testcase(s, n):
    gap = 0
    for b in s:
        if b == "(":
            gap += 1
        elif b == ")":
            gap -= 1
    if gap == 0:
        return "YES"
    return "NO"


def solution():
    t = int(input())
    for _ in range(t):
        n = int(input())
        s = input()
        answer = solve_testcase(s, n)
        print(answer)


solution()

# input
# 1
# 2
# ()
```

## C. Snowfall (00:46)

`-` 이 문제 보고 문제들의 형식이 되게 유사하다고 느꼈다 (테스트 케이스 + $n$의 총 합계 제한)

`-` 생각하는 데 조금 오래 걸렸다. $f(a)$를 최소화한다고 생각하지 말고 결괏값이 $6$의 배수가 아닌 $(l,r)$ 쌍을 최대화한다 생각하고 풀었다

`-` $6$의 배수는 한 쪽 끝에 배치하자

`-` $6$의 배수가 아닌 $2$와 $3$의 배수는 서로 가깝게 있으면 안 된다

`-` 먼저 $6$의 배수를 왼쪽에 배치한 뒤 $2$의 배수를 왼쪽에 배치히자 (순서 바뀌면 손해임)

`-` 그 후 $2$와 $3$의 배수가 아닌 수를 이어서 배치하고 마지막으로 $3$의 배수를 배치하면 된다

```python
def solve_testcase(a):
    six = []
    two = []
    three = []
    other = []
    for a_i in a:
        if a_i % 6 == 0:
            six.append(a_i)
        elif a_i % 2 == 0:
            two.append(a_i)
        elif a_i % 3 == 0:
            three.append(a_i)
        else:
            other.append(a_i)
    result = []
    for array in [six, two, other, three]:
        result.extend(array)
    return result


def solution():
    t = int(input())
    for _ in range(t):
        n = int(input())
        a = list(map(int, input().split()))
        answer = solve_testcase(a)
        print(*answer)


solution()

# input
# 1
# 6
# 1 2 3 5 7 11
```

## D. Palindromex (01:37)

`-` 매내처랑 머지 소트 트리 쓰면 풀 수 있을 것 같은데 일단 다른 인사이트를 얻어 보자

`-` $0$을 포함하는 팰린드롬이 없으면 $\operatorname{mex}$의 최댓값은 $1$이다. 따라서 무조건 $0$을 팰린드롬에 포함시켜야 한다

`-` $0$이 팰린드롬의 중심이 될 수도 있고 아니라면 중심을 기준으로 양 쪽에 각각 등장할 수도 있다

`-` 각 경우의 수마다 최대 팰린드롬을 구한 뒤 이를 단순 순회하면서 $\operatorname{mex}$의 최댓값을 구하면 된다

```python
def maximize_zero_center_pailndrome(i_start, i_end, a):
    n = len(a)
    if a[i_start] != a[i_end]:
        return i_start, i_start
    while i_start > 0 and i_end < n - 1 and a[i_start - 1] == a[i_end + 1]:
        i_start -= 1
        i_end += 1
    return i_start, i_end


def maximize_non_zero_center_pailndrome(zero_indices, a):
    n = len(a)
    left, right = zero_indices
    gap = right - left
    m = (left + right) // 2
    if gap % 2 == 0:
        i_start, i_end = maximize_zero_center_pailndrome(m, m, a)
    else:
        i_start, i_end = maximize_zero_center_pailndrome(m, m + 1, a)
    return i_start, i_end


def mex(i_start, i_end, a):
    palindrome = a[i_start : i_end + 1]
    set_ = set(palindrome)
    mex_ = 0
    while mex in set_:
        mex_ += 1
    return mex_


def solve_testcase(a):
    zero_indices = [i for i, a_i in enumerate(a) if a_i == 0]
    answer = 1
    for i in zero_indices:
        i_start, i_end = maximize_zero_center_pailndrome(i, i, a)
        result = mex(i_start, i_end, a)
        answer = max(result, answer)
    i_start, i_end = maximize_non_zero_center_pailndrome(zero_indices, a)
    result = mex(i_start, i_end, a)
    answer = max(result, answer)
    return answer


def solution():
    t = int(input())
    for _ in range(t):
        n = int(input())
        a = list(map(int, input().split()))
        answer = solve_testcase(a)
        print(answer)


solution()

# input
# 6
# 4
# 1 2 0 3 3 0 2 1
# 2
# 0 1 0 1
# 2
# 1 1 0 0
# 3
# 2 0 2 1 1 0
# 4
# 0 1 3 0 3 1 2 2
# 3
# 0 1 2 1 0 2
```

## E. It All Went Sideways (02:24)

`-` 가장 오른쪽 열에 존재하는 큐브는 *move* 하지 않음에 집중하자

`-` 열에 존재하는 큐브의 개수를 열의 높이라고 할 때 가장 오른쪽 열의 높이를 $h_{\operatorname{base}}$로 잡고 왼쪽으로 순회할 것이다

`-` 만약 현재 열의 높이 $h$가 $h_{\operatorname{base}}$보다 크거나 같다면 $h-h_{\operatorname{base}}$개의 큐브는 *move* 한다

`-` 그렇지 않고 $h$가 $h_{\operatorname{base}}$보다 작다면 기준이 바뀌어 $h_{\operatorname{base}}$는 $h$가 된다

`-` 이제 임의의 열의 높이를 $1$ 차감해서 전체 *move* 횟수를 늘릴 수 있는지 판단해야 한다

`-` 중력이 바뀐뒤에 큐브를 보면 열의 높이를 기준으로 오름차순 정렬되어 있을 것이다. $i$번째 열에서 큐브를 제거했다고 해보자. 그럼 $1\sim i-1$번째 열 중 높이가 $h_i$와 같은 열만큼 무빙한다

`-` 즉, 열의 높이가 같은 것끼리 모은 뒤 가장 큰 집합의 크기에서 $1$을 차감한 만큼 큐브가 *move* 하게 된다

`-` 구현하고 보니 어렵지 않았는데 아이디어를 떠올리는 데 조금 오래 걸렸다

`-` 이 정도 난이도의 문제를 많이 풀어봐야 겠다 (평소에 골드 $1$ 랜디만 할 게 아니라 골랜디를 했어야 됐음)

```python
from collections import defaultdict


def solve_testcase(a):
    num_moves = 0
    h_base = a[-1]
    counter = defaultdict(int)
    counter[h_base] = 1
    for h in a[::-1][1:]:
        if h >= h_base:
            num_moves += h - h_base
        else:
            h_base = h
        counter[h_base] += 1
    return num_moves + max(counter.values()) - 1


def solution():
    t = int(input())
    for _ in range(t):
        n = int(input())
        a = list(map(int, input().split()))
        answer = solve_testcase(a)
        print(answer)


solution()


# input
# 1
# 1
# 1
```

## A. Iskander and Drawings (00:07)

`-` 지문 해석 제대로 못해서 푸는 데 오래 걸렸다 (하나의 선분만 골라서 진행하는 줄 알았다)

`-` 매초 존재하는 모든 선분의 길이가 양쪽에서 $1\operatorname{cm}$씩 줄어든다. 만약 현재 선분의 길이가 $1\operatorname{cm}$ 또는 $2\operatorname{cm}$라면 $1$초 후 해당 선분은 사라진다

`-` 따라서 가장 긴 선분의 길이를 $x$라 할 때 모든 선분이 사라지는 데 걸리는 시간은 $\left\lfloor\frac{x + 1}{2}\right\rfloor$이다

```python
def solve_testcase(string):
    max_length = 0
    curr_length = 0
    for s in string:
        if s == "#":
            curr_length += 1
            max_length = max(curr_length, max_length)
        else:
            curr_length = 0
    return (max_length + 1) // 2


def solution():
    t = int(input())
    for _ in range(t):
        n = int(input())
        s = input().rstrip()
        answer = solve_testcase(s)
        print(answer)


solution()

# input
# 1
# 7
# #######
```

## B. Nikita and Books (00:27)

`-` 이 문제도 지문 해석을 이상하게 했다...

`-` 배열을 오름차순으로 만들 수 있는지 판단하는 문제이다. $i$번째 스택에서 책 $1$권을 빼서 $i+1$번째 스택 위에 놓는다. 그럼 $a_i$는 $1$ 줄고 $a_{i+1}$은 $1$ 증가한다

`-` $i$번째 스택은 $i+1$번째 스택에 최대 몇 권의 책을 줄 수 있을까? 우선 배열이 오름차순이어야 하니 $a_i \le a_{i-1}$이라면 배열을 *neat*로 만들 수 없다. 그렇지 않고 $a_i > a_{i-1}$이라고 하자. 그럼 최대 $a_i - a_{i - 1} - 1$권의 책을 $i+1$번째 스택에 줄 수 있다. 오름차순만 유지하면 되니 책을 최대한으로 줘야 이득이다. $n-1$번째 스택까지 이를 성공적으로 수행한 뒤 $a_{n} > a_{n-1}$을 만족한다면 배열을 *neat*로 만들 수 있다

`-` 배열 맨 앞에 $0$을 추가하면 문제를 더 쉽게 풀 수 있다

```python
def solve_testcase(array):
    n = len(array)
    for i in range(1, n - 1):
        if array[i] <= array[i - 1]:
            return "NO"
        array[i + 1] += array[i] - array[i - 1] - 1
        array[i] = array[i - 1] + 1
    if array[n - 1] <= array[n - 2]:
        return "NO"
    return "YES"


def solution():
    t = int(input())
    for _ in range(t):
        n = int(input())
        a = [0]
        a.extend(map(int, input().split()))
        answer = solve_testcase(a)
        print(answer)


solution()

# input
# 1
# 4
# 1 1 3 5
```

## C. Stepan and Permutation (00:38)

`-` 엄밀하게 증명해서 푼 건 아니다. 추후 에디토리얼을 확인하자

`-` $g = \gcd(x, y)$라고 하자. $g=1$이면 모든 원소를 원하는 순서로 배치할 수 있다. 그렇지 않다면 인덱스 $i$를 $g$로 나눈 나머지에 따라 그룹을 만들자. 해당 그룹 내에선 자유롭게 원소를 배치할 수 있다. 그리고 순서대로 인덱스를 순회하며 해당 그룹 내 원소 중 가장 작은 것을 뽑자. 이것이 이전에 뽑힌 원소보다 작다면 오름차순으로 정렬할 수 없다. 순회를 문제 없이 마치면 배열을 오름차순으로 정렬할 수 있는 것이다

```python
import math


def solve_testcase(array, x, y):
    n = len(array)
    g = math.gcd(x, y)
    if g == 1:
        return "YES"
    groups = [[] for _ in range(g)]
    for i, a_i in enumerate(array):
        groups[i % g].append(a_i)
    for group in groups:
        group.sort(reverse=True)
    prev = 0
    for i in range(n):
        x = groups[i % g].pop()
        if x <= prev:
            return "NO"
        prev = x
    return "YES"    


def solution():
    t = int(input())
    for _ in range(t):
        n, x, y = map(int, input().split())
        p = list(map(int, input().split()))
        answer = solve_testcase(p, x, y)
        print(answer)


solution()

# input
# 1
# 5 2 3
# 5 4 3 2 1
```

## D. Yaroslav and Productivity (01:46)

`-` $i$번째 생산성을 고려하자. 만약 $i$가 $b$에 존재하지 않는다면 그저 $a_i$를 *total productivity*에 합산하는 방법밖에 없다. 그렇지 않다면 기존대로 하거나 $a_1, \dots a_i$에 $-1$을 곱해서 *total productivity*에 기여하는 경우를 고려할 수 있다

`-` 영향력이 $i-1$ 이하인 게시물의 독서 여부가 결정됐다고 했을 때 $a_1$부터 $a_{i-1}$까지 *total productivity*에 기여하는 정도를 양수와 음수로 나누어 고려하자. 만약 영향력이 $i$인 게시물을 읽지 않는다면 양수가 커야 *total productivity*가 커진다. 그렇지 않다면 생산성에 $-1$을 곱할 것이므로 음수가 작아야(절댓값이 커야) *total productivity*가 커진다

`-` 따라서 가장 큰 양수와 그때의 가장 큰 음수, 그리고 가장 작은 음수와 그때의 가장 작은 양수를 기록해두면 이를 바탕으로 상태 전이를 수행할 수 있다

`-` $i$가 $b$에 존재하냐, $a_i$가 양수냐 음수냐에 따라 상태 전이가 달라져서 고려할 경우가 적지 않다. 실제 콘테스트 진행 중엔 이를 처리하는 게 까다롭게 느껴졌다 (푸는 데 너무 오래 걸림)

```python
def solve_testcase(a, b):
    n = len(a)
    b = set(map(str, b))
    dp_prev = [0] * 4  # [max_pos, max_neg, min_pos, min_neg]
    for i in range(1, n):
        dp_curr = dp_prev[:]
        a_i = a[i]
        if a_i >= 0:
            dp_curr[0] += a_i
            dp_curr[2] += a_i
        else:
            dp_curr[1] += a_i
            dp_curr[3] += a_i
        if str(i) not in b:
            dp_prev = dp_curr
            continue
        if a_i >= 0:
            if -dp_prev[3] >= dp_curr[0]:
                dp_curr[0] = -dp_prev[3]
                dp_curr[1] = -(a_i + dp_prev[2])
            if -dp_prev[1] <= dp_curr[2]:
                dp_curr[2] = -dp_prev[1]
                dp_curr[3] = -(a_i + dp_prev[0])
        else:
            if -(a_i + dp_prev[3]) >= dp_curr[0]:
                dp_curr[0] = -(a_i + dp_prev[3])
                dp_curr[1] = -dp_prev[2]
            if -(a_i + dp_prev[1]) <= dp_curr[2]:
                dp_curr[2] = -(a_i + dp_prev[1])
                dp_curr[3] = -dp_prev[0]
        dp_prev = dp_curr
    return dp_prev[0] + dp_prev[1]


def solution():
    t = int(input())
    for _ in range(t):
        n, m = map(int, input().split())
        a = [0]
        a.extend(map(int, input().split()))
        b = list(map(int, input().split()))
        answer = solve_testcase(a, b)
        print(answer)


solution()

# input
# 1
# 5 3
# -1 2 -3 4 -5
# 1 5 3
```

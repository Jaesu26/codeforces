## A. You Delete, I Delete (00:04)

`-` Alice 입장에서 $s$를 사전순으로 가장 크게 만들려면 가장 먼저 등장하는 $\textsf{0}$을 없애야 한다

`-` Bob 입장에서 $s$를 사전순으로 가장 작게 만들려면 가장 먼저 등장하는 $\textsf{1}$을 없애야 한다

```python
def solve_testcase(s):
    s = list(s)
    s.remove("0")
    s.remove("1")
    s = "".join(s)
    return s


def solution():
    t = int(input())
    for _ in range(t):
        s = input().rstrip()
        answer = solve_testcase(s)
        print(answer)


solution()

# input
# 1
# 11001
```

## B. Merge to Match (01:24)

`-` 푸는 데 굉장히 오래 걸렸는데 순수하게 문제가 너무 어려웠다

`-` 배열 $a$의 원소 $n$개와 $b$의 원소 $m$개에 대해 $n+m$개의 원소는 서로 다름을 인지하고 있자. 푸는 도중 까먹어서 $a,b$의 동일한 원소를 제거하는 코드를 작성하는 뻘짓을 해버렸다 (너무 오래 걸림 + 안 풀려서 C번도 시도하며 바꿔가며 풂)

`-` $a$의 원소 $2$개를 사용해 $b$의 원소 하나를 만들 수 있다. 따라서 $n < 2m$이라면 $a$를 $b$로 만들 수 없다

`-` $a$와 $b$를 정렬하자. 배열 $a$에 $b_i$보다 작은 원소가 $j$개 있다고 하자. 만약 $i > j$라면 정답은 NO이다. 왜냐하면 $x\le b_i \le y$를 만족하는 배열 $a$의 $x,y$를 사용해 $b_i$를 만들어야 하는데 이를 만족하는 $x$가 남아나지 않기 때문이다. $b_1$부터 $b_{i-1}$을 만드는 데 이미 $i-1$개를 사용했므으로 적어도 $i$개는 있어야 한다

`-` 마찬가지로 $y$의 개수가 충분한지 따져보자. 배열 $a$에는 $b_i$보다 큰 원소가 $n-j$개 있다. 그리고 배열 $b$에는 $b_i$보다 큰 원소가 $m - i$개 있다. 따라서 $b_i$를 만들기 위해 $n - j \ge m - i + 1$를 만족해야 한다. 그렇지 않다면 정답은 NO이다

`-` 모든 순회를 문제없이 마치면 정답은 YES이다

`-` 정답을 맞히긴 했는데 사실 이게 충분조건인지 잘 모르겠다 (필요조건은 맞다)

`-` 예컨대 $a = \{1, 3, 5, 6\}, b = \{2, 4\}$라고 하자. 일단 주어진 조건은 충족한다

`-` 그런데 $1$과 $3$을 이용해 $2$를 만들면 $4$를 만들 수 없다. 따라서 무지성으로 변환하면 안 되고 잘 생각해서 진행해야 된다. 이때 단순히 조건만 충족한다고 해서 항상 $a$를 $b$로 변환 가능한 경우가 있는 건지는 잘 모르겠다  

```python
def solve_testcase(a, b):
    n, m = len(a), len(b)
    a.sort()
    b.sort()
    if n < 2 * m:
        return "NO"
    j = 0
    for i, b_i in enumerate(b, start=1):
        while j < n and a[j] < b_i:
            j += 1
        if i > j or m - i + 1 > n - j:
            return "NO"
    return "YES"


def solution():
    t = int(input())
    for _ in range(t):
        n, m = map(int, input().split())
        a = list(map(int, input().split()))
        b = list(map(int, input().split()))
        answer = solve_testcase(a, b)
        print(answer)


solution()

# input
# 2
# 4 2
# 1 4 7 9
# 3 10
# 4 2
# 1 3 5 6
# 2 4
```

## C. Maximize the Score (01:39)

`-` $\operatorname{dp}[i]$를 $a$의 $i$번째 인덱스까지 고려했을 때 얻을 수 있는 점수의 최댓값이라 하자. 그리고 $x = a_i$라고 했을 때 $x$가 등장하는 두 인덱스를 $l,r$이라 하자

`-` $i=l$ 또는 $i=r$일텐데 $i=l$이라면 하나의 $a_i$만 홀로 존재한다. 따라서 $x$는 한 번만 등장하는 것이고 구간의 길이는 $1$이니 $\operatorname{dp}[i] = \operatorname{dp}[i - 1] + 1$이다

`-` 이제 $i=r$인 경우를 고려하자. $x$를 연산에 사용한다면 $\operatorname{dp}[i] = \operatorname{dp}[l - 1] + (i - l + 1)^2$이다

`-` 사용하지 않는다면 $a_l$은 다른 원소의 연산에 의해 사라지게 되므로 하나의 $x$만 남게 된다. 따라서 $\operatorname{dp}[i] = \operatorname{dp}[i - 1] + 1$이며 두 경우 중 최댓값을 선택하면 된다

`-` 만약 $a_l$이 다른 원소의 연산에 의해 사라지지 않는다고 해보자. 즉, 연산을 적용한 다른 원소 $y$가 존재하고 두 $y$ 사이에 $a_l$이 존재하지 않는다는 것이다. 근데 이러면 $y$를 연산에 사용하지 않은채로 $x$를 연산에 사용하는 것이 항상 이득이다. 왜냐하면 점수 공식의 특성 때문인데 임의의 자연수 $a,b$에 대해 $(a + b)^2 > a^2 + b^2$이기 때문이다

`-` 따라서 $i=r$인 경우 $\operatorname{dp}[i] = \max(\operatorname{dp}[i - 1] + 1, \operatorname{dp}[l - 1] + (i - l + 1)^2$이다

`-` 구현의 편의성을 위해 $a$ 앞에 $0$을 추가했다

```python
def solve_testcase(array):
    n = len(array)
    left_indices = [None] * (n + 1)
    for i, a_i in enumerate(array):
        if left_indices[a_i] is not None:
            continue
        left_indices[a_i] = i
    dp = [0] * (n + 1)
    for i in range(1, n):
        x = array[i]
        l = left_indices[x]
        if l >= i:
            dp[i] = dp[i - 1] + 1
            continue
        dp[i] = max(dp[i - 1] + 1, dp[l - 1] + (i - l + 1)**2)
    return max(dp)


def solution():
    t = int(input())
    for _ in range(t):
        n = int(input())
        a = [0]
        a.extend(map(int, input().split()))
        answer = solve_testcase(a)
        print(answer)


solution()

# input
# 1
# 3
# 1 1 2 3 3 2
```

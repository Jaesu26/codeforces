## A. Riptide (00:03)

`-` $f(a,b,c)$를 세 사람의 토큰 수가 $a,b,c$일 때 라운드 진행 횟수라고 하자. $a=b$ or $b=c$ or $c=a$라면 게임이 끝나므로 $f(a,b,c,) = 0$이다. 그렇지 않으면 라운드가 진행되므로 $a < b< c$라 할 때 $f(a,b,c) = f(a + 1, b, c - 1) + 1$이다

```python
def f(a, b, c):
    if a == b or b == c or c == a:
        return 0
    a, b, c = sorted([a, b, c])
    return 1 + f(a + 1, b, c - 1)


def solve_testcase(a, b, c):
    return f(a, b, c)


def solution():
    t = int(input())
    for _ in range(t):
        a, b, c = map(int, input().split())
        answer = solve_testcase(a, b, c)
        print(answer)


solution()

# input
# 1
# 4 6 1
```

`-` 에디토리얼 봤다. $a,b,c$ 범위가 작아 시뮬레이션을 돌려도 넉넉하게 통과하므로 깊게 생각하지 않았다

`-` $a< b<c$라 할 때 $b$는 변하지 않는다. 따라서 $a=b$ 또는 $c=b$일 때 게임이 끝나므로 정답은 $\min(b-a, c-b)$이다

## B. Evanescent (00:20)

`-` $s$를 동일한 문자를 가진 연속 블럭으로 나누자. 이때 길이가 $2$ 이상인 블럭에서 문자 $1$개를 삭제하는 건 $|f(s)|$를 줄이지 못한다. 왜냐하면 함수 $f$는 각 블럭마다 하나의 문자만 남기므로 블럭의 길이가 $1$ 이상이라면 결과는 동일하기 때문이다. 따라서 길이가 $1$인 블럭에서 문자를 삭제하도록 하자

`-` 그렇게 해서 고른 문자를 $s_i$라 하자. 기본적으로 블럭 하나가 사라지므로 $|f(s')| = |f(s)| - 1$이다. 만약 $s_{i-1}$과 $s_{i+1}$이 동일하다면 $s_i$를 삭제함으로써 $s_{i-1}$이 속한 블럭과 $s_{i+1}$이 속한 블럭을 병합할 수 있다. 그러면 블럭이 기존보다 $1$개 더 줄어들므로 $|f(s')| = |f(s)| - 2$가 된다

`-` 처음엔 $|f(s')| = |f(s)| - 2$인 경우를 고려하지 못하고 틀려서 AC를 받는 데 오래 걸렸다

```python
def solve_testcase(s):
    n = len(s)
    f_s = 1 + sum(s[i] != s[i + 1] for i in range(n - 1))
    diff = 0
    for i in range(1, n - 1):
        s_i = s[i]
        if s[i - 1] != s_i and s[i + 1] != s_i:
            if s[i - 1] == s[i + 1]:
                diff = 2
                break
            diff = 1
    return f_s - diff


def solution():
    t = int(input())
    for _ in range(t):
        n = int(input())
        s = input().rstrip()
        answer = solve_testcase(s)
        print(answer)


solution()

# input
# 1
# 5
# abbab
```

## C1. Marenol (easy version) (01:04)

`-` 연산의 본질을 깨닫는 데 너무 오래 걸렸다. 문제에선 부분 문자열이 $\textsf{001}, \textsf{100}, \textsf{110}, \textsf{011}$인 경우에만 연산을 적용할 수 있다고 했지만 사실 아니다. 길이가 $3$인 모든 부분 문자열에 연산을 적용할 수 있다. 단, 연산 적용 전과 후가 같아서 의미가 없을 뿐이다

`-` 연산을 적용해도 $\textsf{1}$의 개수는 변하지 않음에 집중하자. $a$를 $b$로 변환하는 건 $a$에 존재하는 $\textsf{1}$의 위치와 $b$에 존재하는 $\textsf{1}$의 위치를 동일하게 만드는 것과 같다

`-` 따라서 $\textsf{1}$을 포함하는 길이가 $3$인 부분 문자열에 연산을 적용하는 걸 고려하자. $\textsf{1}$이 가운데에 있으면 의미가 없다. 양 끝 문자가 $\textsf{1}$이어도 의미가 없다. 따라서 양 끝 문자 중 하나만 $\textsf{1}$이어야 한다. 그리고 연산 적용 결과는 $\textsf{1}$의 위치를 $2$칸 옮기는 것과 같다

`-` 이는 $\textsf{1}$의 인덱스의 홀짝성을 바꾸지 않으며 홀짝성이 같은 인덱스 어디로든 $\textsf{1}$을 옮길 수 있다는 것이다. 그러므로 $a$와 $b$에 대해 홀짝성에 따른 $\textsf{1}$의 개수가 동일할 때만 $a$를 $b$로 변환할 수 있다

```python
def solve_testcase(a, b):
    n = len(a)
    a_even = sum(a[i] == "1" for i in range(n) if i % 2 == 0)
    a_odd = sum(a[i] == "1" for i in range(n) if i % 2 == 1)
    b_even = sum(b[i] == "1" for i in range(n) if i % 2 == 0)
    b_odd = sum(b[i] == "1" for i in range(n) if i % 2 == 1)
    if a_even == b_even and a_odd == b_odd:
        return "YES"
    return "NO"


def solution():
    t = int(input())
    for _ in range(t):
        n = int(input())
        a = input().rstrip()
        b = input().rstrip()
        answer = solve_testcase(a, b)
        print(answer)


solution()

# input
# 1
# 7
# 1001100
# 0000111
```

## C2. Marenol (hard version) (01:12)

`-` C1 문제를 풀었다면 쉽게 해결할 수 있다. 홀짝성에 따라 $a,b$의 $\textsf{1}$의 인덱스를 정렬한 뒤 앞에서부터 매칭시키자. 인덱스 차이가 $x$일 때 필요한 연산 횟수는 $\frac{x}{2}$이다

```python
def solve_testcase(a, b):
    n = len(a)
    a_even = [i for i in range(n) if i % 2 == 0 and a[i] == "1"]
    a_odd = [i for i in range(n) if i % 2 == 1 and a[i] == "1"]
    b_even = [i for i in range(n) if i % 2 == 0 and b[i] == "1"]
    b_odd = [i for i in range(n) if i % 2 == 1 and b[i] == "1"]
    if len(a_even) != len(b_even) or len(a_odd) != len(b_odd):
        return -1
    n_ops = 0
    for i, j in zip(a_even, b_even):
        n_ops += abs(i - j)
    for i, j in zip(a_odd, b_odd):
        n_ops += abs(i - j)
    n_ops //= 2
    return n_ops


def solution():
    t = int(input())
    for _ in range(t):
        n = int(input())
        a = input().rstrip()
        b = input().rstrip()
        answer = solve_testcase(a, b)
        print(answer)


solution()

# input
# 1
# 7
# 1001100
# 0000111
```

## D. Silhouette (Upsolving)

`-` $a_i$의 *shadow*는 배열 $a$에서 $a_i$보다 작은 원소들의 합이다. $a$에서 가장 작은 원소는 자신보다 작은 원소가 존재하지 않으므로 *shadow*가 $0$이다. 따라서 $b$에는 $0$이 존재해야 하며 그렇지 않으면 정답은 $-1$이다. 만약 모든 *shadow*가 $0$이라면 $a$의 모든 원소가 $1$일 때 사전순으로 가장 작다

`-` 한편, 동일한 *shadow*를 가진다면 배열 $a$에서도 동일한 값을 가져야 한다. 그렇지 않다면 더 큰 값이 더 작은 값의 *shadow*에 더해 추가로 상대 원소도 *shadow*에 포함하므로 둘의 *shadow*는 다르게 된다

`-` 이를 바탕으로 $a$를 재구성할 것이다. $b$에서 유니크한 원소만 골라 오름차순 정렬하자. 그리고 각 원소별 빈도수 배열 $c$를 가지고 있자

`-` 시작은 $0$이다. 이것이 $c_1$개 존재하며 $0$ 다음 원소는 $x$라고 하자

`-` *shadow*가 $x$라는 것은 자기보다 작은 원소의 합이 $x$라는 것이다. 이는 $c_1$개 존재하며 이들보다 작은 원소의 합은 $0$이다. 그리고 이들은 모두 같은 값을 가져야 한다. 따라서 *shadow*가 $0$인 원소는 배열 $a$에서 $\frac{x - 0}{c_1}$을 값으로 가지게 된다. 만약 $x$가 $c_1$의 배수가 아니거나 $\frac{x-0}{c_1}$이 여태까지 확정된 배열 $a$의 원소보다 작거나 같다면 모순이다. 이를 모든 *shadow*에 대해서 크기순으로 수행하자

`-` 가능한 $a$ 중 사전순으로 가장 작은 걸 출력해야 되지만 실질적으로 $a$의 원소를 마음대로 선택할 수 있는 건 *shadow*가 가장 클 때 뿐이다. 여태까지 확정된 배열 $a$의 원소 중 최댓값보다 크기만 하면 되므로 이들 중 최솟값인 $\max(a) + 1$로 설정하자. 이렇게 만들어진 $a$는 $b$를 바탕으로 재구성한 배열 중 사전순으로 가장 작게 된다

```python
from itertools import groupby


def solve_testcase(b):
    b = sorted([(x, i) for i, x in enumerate(b)])
    shadows = [[key, [i for _, i in group]] for key, group in groupby(b, lambda e: e[0])]
    if shadows[0][0] != 0:
        return [-1]
    n, m = len(b), len(shadows)
    a = [0] * n
    prev = 0
    for i in range(m - 1):
        diff = shadows[i + 1][0] - shadows[i][0]
        c = len(shadows[i][1])
        k = diff // c
        if diff % c != 0 or k <= prev:
            return [-1]
        for j in shadows[i][1]:
            a[j] = k
        prev = k
    for j in shadows[-1][1]:
        a[j] = prev + 1
    return a


def solution():
    t = int(input())
    for _ in range(t):
        n = int(input())
        b = list(map(int, input().split()))
        answer = solve_testcase(b)
        print(*answer)


solution()

# input
# 1
# 5
# 0 4 0 4 14
```

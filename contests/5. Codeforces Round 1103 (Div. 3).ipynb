## A. Games on the Train (00:03)

`-` $\max(h) - \min(h) + 1$을 출력하면 된다 ($x_i$는 최소 $1$)

```python
def solve_testcase():
    n = int(input())
    h = list(map(int, input().split()))
    k = max(h) - min(h) + 1
    print(k)


def solution():
    t = int(input())
    for _ in range(t):
        solve_testcase()


solution()

# input
# 1
# 1
# 1
```

## B. Tatar TV Show (00:08)

`-` $s_i$가 $0$이면 냅두자. 아니라면 반전시켜야 한다

`-` 이때 $s_{i + k}$가 $0$이라면 $1$이 되고 $1$이라면 $0$이 된다

`-` $s_1$부터 $s_{i-1}$까지 $0$이라고 하자. 이때 조건을 만족하는 선에서 $s_i$를 반전시킬려면 $i$를 선택하는 것 밖에 없다 ($s_{i-k}$는 이미 $0$이라서 $i-k$를 선택하면 조건을 만족시키지 못함)

`-` 따라서 $s_1$부터 $s_{n-k}$까지 순회하며 작업하자

`-` 순회가 끝난 뒤 $s$의 모든 문자가 $0$이면 Yes이고 아니면 No이다

```python
def solve_testcase():
    n, k = map(int, input().split())
    s = list(map(int, list(input().rstrip())))
    for i in range(n - k):
        if s[i] == 1:
            s[i] = 0
            s[i + k] = 1 - s[i + k]
    result = sum(s) == 0
    if result:
        print("YES")
        return
    print("NO")


def solution():
    t = int(input())
    for _ in range(t):
        solve_testcase()


solution()

# input
# 1
# 1 1
# 1
```

## C. Omsk Programmers (00:25)

`-` 무지성으로 구현했다가 시간 초과가 발생해서 $1$번 틀렸다

`-` $a,b$는 대칭성을 띄므로 $a < b$로 가정하자 ($1,2$번 연산이 $a,b$에 대해 동일함)

`-` $f(a,b,x)$를 $a$를 $b$로 만들기 위해 필요한 최소 연산 횟수라고 하자

`-` 만약 $a=b$라면 $0$을 반환해야 한다. $b-a = 1$이라면 $1$을 반환하면 된다

`-` 그렇지 않다면 $g = b - a$라고 하자. $a$에 $1$번 연산을 $g$번 적용해서 둘을 같게 만들 수 있다

`-` 아니라면 $2$번 연산을 $b$에 적용해야 한다

`-` 두 경우 중 최솟값을 반환하면 되므로 $f(a,b,x) = \min\left(b - a,\; f\left(a, \left\lfloor\frac{b}{x}\right\rfloor, x\right) + 1\right)$이다

```python
def f(a, b, x):
    if a > b:
        a, b = b, a
    if a == b:
        return 0
    if b - a == 1:
        return 1
    return min(b - a, f(a, b // x, x) + 1)


def solve_testcase():
    a, b, x = map(int, input().split())
    result = f(a, b, x)
    return result


def solution():
    t = int(input())
    for _ in range(t):
        answer = solve_testcase()
        print(answer)


solution()

# input
# 1
# 1 1 1
```

## D. Brand New Tatar TV Show (00:51)

`-` Arseniy가 첫 수를 둔 다음 Egor랑 Dabir가 번갈아서 한다

`-` Arseniy가 첫 수를 두어서 Egor가 이기게 해야 된다

`-` 문제 조건에 따라 $k$ 이하로 차이나는 원소들을 한 그룹으로 묶자

`-` 그럼 그룹간의 원소 선택은 불가능하다 (차이가 $k$를 초과함)

`-` 임의의 그룹에 대해 가장 큰 원소를 $x$라 하고 $c$개 있다고 해보자

`-` $c$가 짝수라면 Arseniy가 $x$를 고를 시 $x$는 홀수 개만 남아있으며 Egor랑 Dabir는 $x$만 선택할 수 있다

`-` 그런데 $x$는 홀수 개만 남아있으므로 Egor가 이기게 된다

`-` 만약 그룹 내 모든 원소가 $x$이고 $c$가 홀수라면 Egor를 이기게 할 수 없으므로 다른 그룹에서 선택을 강구해야 한다

`-` 그렇지 않다면 $x$ 다음으로 큰 원소를 Arseniy가 고른 다음 Egor가 $x$를 고르면 된다

`-` 즉, 그룹 내 모든 원소가 $x$이고 $c$가 홀수인 경우만 아니라면 Arseniy는 Egor를 이기게 만들 수 있다

`-` 모든 그룹을 순회하며 적어도 하나의 그룹에 대해 Egor를 이기게 만들 수 있으면 된다

```python
def divide(array, k):
    groups = []
    g = []
    last = array[0]
    for a in array:
        if a - last <= k:
            g.append(a)
        else:
            groups.append(g)
            g = [a]
        last = a
    groups.append(g)
    return groups


def can_win(group):
    n = len(group)
    x = max(group)
    i = group.index(x)
    if i == 0:
        return n % 2 == 0
    return True


def solve_testcase():
    n, k = map(int, input().split())
    a = list(map(int, input().split()))
    a.sort()
    groups = divide(a, k)
    answer = "NO"
    for g in groups:
        if can_win(g):
            answer = "YES"
            return answer
    return answer


def solution():
    t = int(input())
    for _ in range(t):
        answer = solve_testcase()
        print(answer)


solution()

# input
# 1
# 1 1
# 1
```

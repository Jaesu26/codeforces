## A. farmpiggie and Subset Sum (00:08)

`-` 진짜 모르겠다. A번인데 대충 때려 맞추면 되지 않을까?

`-` $1$을 맨 마지막에 배치하고 앞에서부터 오름차순으로 배치하면 될 것 같다 (아마도)

```python
def solve_testcase(n):
    return list(range(2, n + 1)) + [1]


def solution():
    t = int(input())
    for _ in range(t):
        n = int(input())
        answer = solve_testcase(n)
        print(*answer)


solution()

# input
# 1
# 2
```

`-` 에디토리얼이 아름답다

`-` $n$은 짝수이다. 홀수번째 인덱스에 짝수를 배치하고 짝수번째 인덱스에 홀수를 배치한다면 $c_{\operatorname{final}}$은 짝수가 된다

`-` 따라서 $c_{\operatorname{final}}$은 홀수인 $1$이 될 수 없다

`-` $1$뿐만 아니라면 홀수면 불가능하다. 나는 A번 문제이니까 쉽다 생각해 대충 때려 맞추려 해서 이런식으로 확장시키지 못했다

`-` 솔직히 다시 풀어도 생각 못할 것 같다. 이를 파악하려면 $n$이 짝수인 걸 시작으로 무언가(대칭성?) 있지 않을까 탐구해야 될 것 같다 (대충 때려 맞추는 게 정답이라면 굳이 $n$이 짝수일 필요가 없다)

`-` Good Problem!

## B. ezraft and Array (00:30)

`-` 이거 보니까 구성적 문제는 때려 맞추려고 하는 특성이 강하다

`-` 생각을 좀 오래했다

`-` 먼저 $n=1$이면 정답은 $1$로 하자 (뭐든 상관 없다)

`-` $n=2$라면 해가 존재하지 않으므로 $-1$이다. 이유는 다음과 같다

`-` 배열을 구성하는 두 원소를 $a_1,a_2$라고 하자. 문제의 조건에 따라 $a_1 + a_2 = ka_1 = ta_2,\, (t,k > 1)$를 만족한다

`-` 이를 정리하면 $a_2 = a_1(k - 1)$이고 $a_1 = a_2(t-1)$이 된다. 이렇게 되면 $a_1$은 $a_2$의 배수이면서 약수가 된다. 그럼 둘은 같아야 하는데 그렇게 되면 배열의 원소는 서로 다르다는 조건과 모순된다. 따라서 $n=2$일 땐 조건을 만족하는 배열이 존재하지 않는다

`-` 이제 $n > 2$라고 가정하고 문제를 풀겠다

`-` $s = \sum\limits_{i=1}^{n}a_i$로 정의하자. 만약 $s$가 짝수라면 반으로 나눌 수 있다

`-` 근데 배열의 원소는 서로 달라야 한다. $\frac{s}{2}$를 짝수라 가정하고 또 반으로 나누자. 한 번 반으로 나눌 때마다 배열의 길이가 $1$ 증가한다

`-` 근데 단순히 이것만 반복해서는 서로 다른 원소여야 한다는 조건을 맞추지 못한다. 마지막엔 반이 아니라 $\frac{1}{3}$과 $\frac{2}{3}$로 나누자

`-` 그럼 모든 원소는 서로 다르며 각 원소는 $s$의 약수이니 문제의 정답이 된다

`-` 이를 실현하면 $s=3\cdot 2^r$꼴이 되는데 반으로 나누다가 $6$이 되면 $2$와 $4$로 나눌 것이다

`-` 그럼 배열의 원소는 $2,4, 6, 12, \dots, 6 \cdot 2^{n-3}$이 되며 $6\cdot 2^{47} \le 10^{17}$이므로 범위를 초과하지도 않으니 문제가 되지 않는다

```python
def solve_testcase(n):
    if n == 1:
        return [1]
    if n == 2:
        return [-1]
    return [2, 4] + [6 * 2**i for i in range(n - 2)]


def solution():
    t = int(input())
    for _ in range(t):
        n = int(input())
        answer = solve_testcase(n)
        print(*answer)


solution()

# input
# 1
# 5
```

`-` 풀고 나서 생각해보니 $a_i$의 범위가 독특하다는 점이 힌트가 될 수 있었을 것 같다 (범위 제한으로 $10^{17}$은 처음 본다)

## C. 0mar and Alternating Sums (01:26)

`-` 배열이 비내림차순으로 정렬되어 있음에 집중하자. 교대합을 $0$으로 만드는 한 가지 방법은 동일한 원소를 짝수 개씩 사용하는 것이다 (번갈아가며 더하고 빼서 상쇄됨)

`-` $m$개의 원소 중 순서에 상관없이 짝수 개를 선택하는 경우의 수는 $2^{m-1}$이다. 홀수 개도 동일하며 검색해서 찾았다 (처음에 경우의 수 잘못 이해해서 삽질함)

`-` 배열에 유니크한 원소가 $k$개 있다고 가정하자

`-` 그럼 임의의 원소 $x_i$가 $c_i$개 존재할 때 교대합이 $0$인 부분 수열의 개수는 $\prod\limits_{i=1}^{k} 2^{c_i - 1}$이다

`-` 다른 방법은 $-1$과 $x_i + 1 = x_{i+1}$을 만족하는 $x_i, x_{i+1}$을 홀수 개 사용하고 나머지 원소들을 짝수 개씩 사용하는 것이다 ($-1$이 최소 $1$개 존재해야 한다)

`-` $x_i + 1 = x_{i+1}$을 만족하는 $i$의 개수가 $r$일 때 교대합이 $0$인 부분 수열의 개수는 $r\prod\limits_{i=1}^{k} 2^{c_i - 1}$이다

`-` 두 경우를 취합하면 정답이다

```python
from collections import Counter


def solve_testcase(array, mod):
    counter = list(Counter(map(str, array)).items())
    counter = [(int(key), value) for key, value in counter]
    counter.sort()
    prod = 1
    for x, c in counter:
        prod *= pow(2, c - 1, mod)
        prod %= mod
    if counter[0][0] != -1:
        return prod
    r = sum(1 for i in range(len(counter) - 1) if counter[i][0] + 1 == counter[i + 1][0])
    return ((r + 1) * prod) % mod


def solution():
    mod = 10**9 + 7
    t = int(input())
    for _ in range(t):
        n = int(input())
        a = list(map(int, input().split()))
        answer = solve_testcase(a, mod)
        print(answer)


solution()

# input
# 1
# 5
# -1 1 1 2 3
```

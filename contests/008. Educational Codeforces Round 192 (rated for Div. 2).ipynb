## A. Bigrams (00:06)

`-` $2$번 이상 등장하는 바이그램이 존재하는지 판단하면 된다

`-` $c_i$가 $3$이상이면 같은 알파벳을 $3$개 이어붙여 $2$개 이어붙인 바이그램 $2$개를 만들 수 있으므로 YES이다

`-` 아니라면 빈도수의 최댓값과 그 다음 최댓값을 따지자. 둘 다 $2$보다 크거나 같으면 이를 번갈아 배치해서 동일 바이그램 $2$개를 만들 수 있다

`-` 그 외의 경우엔 불가능이다

```python
def solve_testcase(array):
    n = len(array)
    array.sort()
    if array[-1] >= 3:
        return "YES"
    if n == 1:
        return "NO"
    m1, m2 = array[-1], array[-2]
    if m1 >= 2 and m2 >= 2:
        return "YES"
    return "NO"


def solution():
    t = int(input())
    for _ in range(t):
        k = int(input())
        c = list(map(int, input().split()))
        answer = solve_testcase(c)
        print(answer)


solution()

# input
# 1
# 3
# 1 2 3
```

## B. Predominant Frequency Division (00:28)

`-` 배열을 세 그룹으로 나눠야 한다. 이때 비어있으면 안 되고 $i$보다 큰 원소의 개수는 그룹 크기의 절반 이하여야 한다

`-` 배열의 원소는 $1,2,3$ 중 하나니까 $i=3$인 경우 빈 그룹만 아니면 항상 조건이 성립된다. 쓰고 남은 것들을 배정하자

`-` $i=1$인 경우 앞에서부터 그룹에 원소를 추가하다가 조건을 만족하면 추가를 그만두자. $i=1$일 때 조건을 만족하면 $i=2$일 때의 조건도 만족하므로 굳이 원소를 더 추가하는 건 $i=2$인 그룹의 구성을 어렵게 할 뿐이다. 단, 그룹 크기가 홀수이면서 조건을 만족하는 경우 다음 원소가 $3$인지 확인하고 $3$이면 추가하자. 이는 $i=2$인 그룹을 구성할 때 이득을 보기 위함이다. 어차피 $3$은 $2$보다도 크니 조건 충족에 방해만 된다

`-` $i=1$인 그룹을 만든 뒤 같은 방법으로 $i=2$인 그룹을 만들자. 그리고 남은 것들을 $i=3$인 그룹에 배정하자

`-` 모든 그룹이 비어있지 않으면 YES이고 아니면 NO이다

`-` 예제가 친절해서 디버깅하기 편했다

```python
def solve_testcase(array):
    n = len(array)
    num_greater_one = 0
    index = n
    for i, a_i in enumerate(array, start=1):
        if a_i > 1:
            num_greater_one += 1
        if num_greater_one <= (i // 2):
            index = i
            break
    if index % 2 == 1 and index < n and array[index] == 3:
        index += 1
    size = 0
    num_greater_two = 0
    is_possible = "NO"
    for i in range(index, n - 1):
        size += 1
        if array[i] > 2:
            num_greater_two += 1
        if num_greater_two <= (size // 2):
            is_possible = "YES"
            break
    return is_possible


def solution():
    t = int(input())
    for _ in range(t):
        n = int(input())
        a = list(map(int, input().split()))
        answer = solve_testcase(a)
        print(answer)


solution()

# input
# 1
# 5
# 1 3 3 1 1
```

## C. Unstable Elements (01:09)

`-` 기존 배열 내 유니크한 원소의 개수를 $q$라 하자. 삭제 연산을 통해 삭제되는 원소의 개수는 $q$이며 복제 연산을 통해 복제되는 원소의 개수도 $q$이다

`-` 기존 배열을 기반으로 정수와 빈도를 원소로 가지는 튜플 배열을 만들자. 이것의 길이가 유니크한 원소의 개수가 될 것이다

`-` 길이가 $k$인 도달 가능한 모든 배열은 유니크한 원소의 개수만 다를 것이다. 임의의 연산 후에도 배열은 정렬되어 있으니 유니크한 원소의 개수가 같은데 서로 다른 배열일 수는 없다 (이에 따라 원소의 종류는 중요하지 않아 빈도만 원소로 가져도 됨)

`-` 그럼 길이가 $k$인 도달 가능한 배열의 유니크한 원소의 개수 $p$는 $1$과 $q$ 사이의 값이 된다. 각 경우에 대해 실현 가능한지 판단하자

`-` 가능한 $p$의 값으로 $q$부터 $1$까지 역순으로 고려할 것이다. 어떤 정수의 최소 빈도를 $x$라 할 때 유니크한 원소의 개수가 변하면 안 되니 삭제 연산은 최대 $x-1$번만 할 수 있으며 복제 연산은 원하는 만큼 가능하다

`-` 현재 배열 내 원소의 개수를 $s$라 할 때 $|s-k|$가 $p$의 배수이면서 $k\le s - p(x - 1)$를 만족해야 한다. 그럼 도달 가능하다

`-` 이제 다른 종류의 도달 가능한 배열을 만들려면 유니크한 원소의 개수를 줄여야 된다. 삭제 연산을 $x$번 진행하자. 그럼 빈도가 $x$인 원소의 수만큼 유니크한 원소의 개수가 줄어들게 되며 배열 내 원소의 개수 또한 $xp$만큼 줄어들게 된다. 그리고 이에 대해 위의 과정을 반복적으로 수행하면 된다

`-` 삭제 연산을 수행한 후 매번 배열을 순회하며 빈도를 $x$만큼 줄이는 건 비효율적이다. 대신 여태까지 삭제 연산을 몇 번 했고 이에 따라 배열 내 원소의 개수가 얼마나 줄었는지 추적하자

`-` 그럼 $O(n)$에 길이가 $k$인 도달 가능한 배열의 개수를 셀 수 있다

`-` 매번 최소 빈도를 $O(1)$에 찾고자 배열을 정렬할 것이므로 전체 알고리즘의 시간 복잡도는 최악의 경우 $O(n \log n)$이다

```python
from collections import Counter


def solve_testcase(array, k):
    b = list(Counter(map(str, array)).values())
    b.sort(reverse=True)
    count = 0
    num_deletions = 0
    s = len(array)
    while b:
        p = len(b)
        x = b[-1] - num_deletions
        if abs(k - s) % p == 0 and k >= s - p * (x - 1):
            count += 1
        while b and b[-1] - num_deletions == x:
            b.pop()
        num_deletions += x
        s -= x * p
    return count


def solution():
    t = int(input())
    for _ in range(t):
        n, k = map(int, input().split())
        a = list(map(int, input().split()))
        answer = solve_testcase(a, k)
        print(answer)


solution()

# input
# 1
# 10 7
# 1 1 1 2 2 3 3 3 3 4
```

`-` 딕셔너리의 키로 정수가 아닌 문자열을 사용하면 TLE를 방지할 수 있다

`-` 정수는 해시 충돌 때문에 TLE가 발생한다 (저격 테케 존재, 저격하기 쉬운 형태이다)

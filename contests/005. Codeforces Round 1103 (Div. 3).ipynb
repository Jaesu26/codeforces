## B. Tatar TV Show (00:08)

`-` $s_i$가 $0$이면 냅두자. 아니라면 반전시켜야 한다

`-` 이때 $s_{i + k}$가 $0$이라면 $1$이 되고 $1$이라면 $0$이 된다

`-` $s_1$부터 $s_{i-1}$까지 $0$이라고 하자. 이때 조건을 만족하는 선에서 $s_i$를 반전시킬려면 $i$를 선택하는 것 밖에 없다 ($s_{i-k}$는 이미 $0$이라서 $i-k$를 선택하면 조건을 만족시키지 못함)

`-` 따라서 $s_1$부터 $s_{n-k}$까지 순회하며 작업하자

`-` 순회가 끝난 뒤 $s$의 모든 문자가 $0$이면 Yes이고 아니면 No이다

```python
def solve_testcase():
    n, k = map(int, input().split())
    s = list(map(int, list(input().rstrip())))
    for i in range(n - k):
        if s[i] == 1:
            s[i] = 0
            s[i + k] = 1 - s[i + k]
    result = sum(s) == 0
    if result:
        print("YES")
        return
    print("NO")


def solution():
    t = int(input())
    for _ in range(t):
        solve_testcase()


solution()

# input
# 1
# 1 1
# 1
```

## C. Omsk Programmers (00:25)

`-` 무지성으로 구현했다가 시간 초과가 발생해서 1번 틀렸다

`-` $a,b$는 대칭성을 띄므로 $a < b$로 가정하자 ($1,2$번 연산이 $a,b$에 대해 동일함)

`-` $f(a,b,x)$를 $a$를 $b$로 만들기 위해 필요한 최소 연산 횟수라고 하자

`-` 만약 $a=b$라면 $0$을 반환해야 한다. $b-a = 1$이라면 $1$을 반환하면 된다

`-` 그렇지 않다면 $g = b - a$라고 하자. $a$에 $1$번 연산을 $g$번 적용해서 둘을 같게 만들 수 있다

`-` 아니라면 $2$번 연산을 $b$에 적용해야 한다

`-` 두 경우 중 최솟값을 반환하면 되므로 $f(a,b,x) = \min\left(b - a,\; f\left(a, \left\lfloor\frac{b}{x}\right\rfloor, x\right) + 1\right)$이다

```python
def f(a, b, x):
    if a > b:
        a, b = b, a
    if a == b:
        return 0
    if b - a == 1:
        return 1
    return min(b - a, f(a, b // x, x) + 1)


def solve_testcase():
    a, b, x = map(int, input().split())
    result = f(a, b, x)
    return result


def solution():
    t = int(input())
    for _ in range(t):
        answer = solve_testcase()
        print(answer)


solution()

# input
# 1
# 1 1 1
```

## D. Brand New Tatar TV Show (00:51)

`-` Arseniy가 첫 수를 둔 다음 Egor랑 Dabir가 번갈아서 한다

`-` Arseniy가 첫 수를 두어서 Egor가 이기게 해야 된다

`-` 문제 조건에 따라 $k$ 이하로 차이나는 원소들을 한 그룹으로 묶자

`-` 그럼 그룹간의 원소 선택은 불가능하다 (차이가 $k$를 초과함)

`-` 임의의 그룹에 대해 가장 큰 원소를 $x$라 하고 $c$개 있다고 해보자

`-` $c$가 짝수라면 Arseniy가 $x$를 고를 시 $x$는 홀수 개만 남아있으며 Egor랑 Dabir는 $x$만 선택할 수 있다

`-` 그런데 $x$는 홀수 개만 남아있으므로 Egor가 이기게 된다

`-` 만약 그룹 내 모든 원소가 $x$이고 $c$가 홀수라면 Egor를 이기게 할 수 없으므로 다른 그룹에서 선택을 강구해야 한다

`-` 그렇지 않다면 $x$ 다음으로 큰 원소를 Arseniy가 고른 다음 Egor가 $x$를 고르면 된다

`-` 즉, 그룹 내 모든 원소가 $x$이고 $c$가 홀수인 경우만 아니라면 Arseniy는 Egor를 이기게 만들 수 있다

`-` 모든 그룹을 순회하며 적어도 하나의 그룹에 대해 Egor를 이기게 만들 수 있으면 된다

```python
def divide(array, k):
    groups = []
    g = []
    last = array[0]
    for a in array:
        if a - last <= k:
            g.append(a)
        else:
            groups.append(g)
            g = [a]
        last = a
    groups.append(g)
    return groups


def can_win(group):
    n = len(group)
    x = max(group)
    i = group.index(x)
    if i == 0:
        return n % 2 == 0
    return True


def solve_testcase():
    n, k = map(int, input().split())
    a = list(map(int, input().split()))
    a.sort()
    groups = divide(a, k)
    answer = "NO"
    for g in groups:
        if can_win(g):
            answer = "YES"
            return answer
    return answer


def solution():
    t = int(input())
    for _ in range(t):
        answer = solve_testcase()
        print(answer)


solution()

# input
# 1
# 1 1
# 1
```

## E. Friendly Gifts (Upsolving)

`-` 에디토리얼 봤다

`-` Dabir가 고른 *good* 배열 $G_1$과 Egor가 고른 *good* 배열 $G_2$를 합쳐도 *good* 배열이여야 한다

`-` 따라서 $G_1$의 최솟값과 최댓값을 $\min_{G_1}, \max_{G_1}$이라 하고 $G_2$의 최솟값과 최댓값을 $\min_{G_2},\max_{G_2}$라 하자

`-` 일반성을 잃지 않고 $\max_{G_1} < \min_{G_2}$라고 한다면 $\max_{G_1} + 1 = \min_{G_2}$를 만족해야 한다

`-` 두 배열이 동일한 원소를 하나도 가지고 있지 않으므로 두 배열은 서로 겹치지 않는다 (키 포인트)

`-` 모든 *good* 배열 중 적당히 $2$개를 골라 조건을 만족하면 되며 이 중 가장 긴 배열의 길이가 정답이다

`-` 이에 따라 가능한 *good* 배열을 모두 구해두자. 길이가 $k$인 임의의 *good* 배열 $b$는 정렬했을 때 $b_i - b_{i-1} = 1$을 만족한다. 이는 $\min_b + k - 1 = \max_b$이며 $b$의 모든 원소는 서로 다름을 뜻한다

`-` 이를 활용해 임의의 $i,j$에 대해 $a_i \sim a_j$의 *good* 배열 여부를 판단하자

`-` 고정된 $i$에 대해 $j$를 $i$부터 $n$까지 늘려갈 것이다. 만약 여태까지 등장한 원소가 서로 다르고 $\min_{a_i \sim a_j} + j - i = \max_{a_i \sim a_j}$라면 $a_i \sim a_j$는 *good* 배열이다. 만약 이미 등장한 원소가 또 등장하면 $a_i$를 첫 번째 원소로 가지는 *good* 배열은 더 이상 존재하지 않으니 다음 $i$로 넘어가서 탐색을 계속하면 된다

`-` 최솟값, 최댓값, 등장한 원소 목록을 추적하면 고정된 $i$에 대해 $O(n)$에 *good* 배열을 모두 찾을 수 있다. 가능한 $i$는 $n$개이므로 총 $O\left(n^2\right)$에 가능한 모든 *good* 배열을 얻을 수 있다

`-` 최솟값과 최댓값이 $\min, \max$인 *good* 배열이 존재하면 $\operatorname{dp}[\max][\min]$는 $1$이며 아니라면 $0$이라 하자

`-` 가능한 $G_1$의 길이 $k$는 $\left\lfloor\frac{n}{2}\right\rfloor$부터 $1$까지이다. 이에 따라 가능한 $\min_{G_1}$은 $1$부터 $n- 2k + 1$까지이다

`-` 따라서 $\operatorname{dp}[\min_{G_1} + k - 1][\min_{G_1}]\times \operatorname{dp}[\min_{G_1} + 2k - 1][\min_{G_1} + k] = 1$을 만족하는 $k$ 중 최댓값이 정답이 된다

`-` 전체 알고리즘의 시간 복잡도는 $O\left(n^2\right)$이다

```python
def check_good_array(array, i, dp):
    n = len(array)
    mn, mx = n + 1, 0
    seen = [False] * (n + 1)
    for j in range(i, n):
        a_j = array[j]
        if seen[a_j]:
            return
        seen[a_j] = True
        if a_j > mx:
            mx = a_j
        if a_j < mn:
            mn = a_j
        if mn + j - i == mx:
            dp[mx][mn] = 1


def solve_testcase(array):
    n = len(array)
    dp = [[0] * (i + 1) for i in range(n + 1)]
    for i in range(n):
        check_good_array(array, i, dp)
    for k in range(n // 2, 0, -1):
        for mn in range(1, n - 2 * k + 2):
            if dp[mn + k - 1][mn] * dp[mn + 2 * k - 1][mn + k] == 1:
                return k
    return 0


def solution():
    t = int(input())
    for _ in range(t):
        n = int(input())
        array = list(map(int, input().split()))
        answer = solve_testcase(array)
        print(answer)


solution()

# input
# 1
# 2
# 2 1
```

## F1. Elections in Saransk (easy version) (Upsolving)

`-` 에디토리얼 봤다. 어렵지 않은 문제이다

`-` $x=1$이므로 임의의 $i,j$에 대해 $p_i,p_j$는 서로소여야 한다

`-` $2$부터 $\max(a)$까지의 소수 $f_1,f_2,\dots, f_m$을 고려하자

`-` $f_i$의 배수인 $a_i, a_j, \dots, a_k$를 고려하자. $p$들은 서로소여야 하니 이들 중 하나만 $f_i$를 가질 수 있다

`-` $f_i$의 배수가 아닌 $d$에 대해 $b = d\cdot {f_i}^e$라면 $f_i$를 $1$개부터 $e$개까지 취할 수 있다. 그리고 글로벌하게 봤을 때 $f_i$를 아무도 가지지 않을 수도 있다

`-` 이에 따라 함수 $v$를 다음과 같이 정의하자

`-` $v(f_i)=1 + \sum\limits_{i=1}^{n} e_i,\quad \text{단, $f_i \mid a_i$이면 $a_i = d\cdot {f_i}^{e_i},\quad f_i\nmid a_i$이면 $e_i = 0$}$

`-` 각 소수별 경우의 수를 곱해주면 되니 $\prod\limits_{i=1}^{m} v(f_i)$가 정답이 된다

```python
from collections import defaultdict


def find_prime_factors(x):
    prime_factors = defaultdict(int)
    for d in range(2, x + 1):  
        if d * d > x:
            break
        while x % d == 0:
            x //= d
            prime_factors[d] += 1
    if x > 1:
        prime_factors[x] += 1
    return prime_factors


def solve_testcase(array, mod):
    global_prime_factors = defaultdict(int)
    for a in array:
        prime_factors = find_prime_factors(a)
        for p, c in prime_factors.items():
            global_prime_factors[p] += c
    count = 1
    for _, c in global_prime_factors.items():
        count *= (c + 1)
        count %= mod
    return count


def solution():
    mod = 10**9 + 7
    t = int(input())
    for _ in range(t):
        n, x = map(int, input().split())
        array = list(map(int, input().split()))
        answer = solve_testcase(array, mod)
        print(answer)


solution()

# input
# 1
# 4 1
# 1 2 3 4
```

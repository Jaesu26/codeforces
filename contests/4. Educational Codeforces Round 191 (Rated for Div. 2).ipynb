## A. AI Project Development (00:11)

`-` Nikita가 AI를 사용하는 경우와 사용하지 않는 경우를 나누어 계산한 뒤 최솟값을 반환하면 된다

```python
def solve_testcase():
    n, x, y, z = map(int, input().split())
    h_no = n // (x + y) if n % (x + y) == 0 else n // (x + y) + 1
    h_yes = z + (n - x * z) // (x + 10 * y) if (n - x * z) % (x + 10 * y) == 0 else z + (n - x * z) // (x + 10 * y) + 1
    h_yes = max(0, h_yes)
    h = min(h_no, h_yes)
    print(h)


def solution():
    t = int(input())
    for _ in range(t):
        solve_testcase()


solution()

# input
# 1
# 1 1 1 1
```

## B. Different Distances (00:29)

`-` $n$이 $2$ 또는 $3$인 경우는 하드 코딩했다

`-` $n$이 짝수라면 $1,1,2,2,\dots, n,n, 1, 2 ,1, 2, \dots, n-1, n, n-1,n$을 출력하자

`-` $n$이 홀수라면 $1,1,2,2,\dots, n,n, n-1,n-1, 1, 2 ,\dots, n-1,n ,1, 2 ,\dots, n-1,n$을 출력하자

```python
def solve_testcase():
    n = int(input())
    if n == 2:
        print(2, 1, 1, 2, 1, 2, 2, 1)
        return
    if n == 3:
        print(1, 1, 2, 1, 2, 3, 1, 3, 2, 2, 3, 3)
        return
    if n % 2 == 0:
        array = []
        for i in range(1, n + 1):
            array.extend([i, i])
        for i in range(1, n + 1, 2):
            array.extend([i, i + 1, i, i + 1])
        print(*array)
        return
    array = []
    for i in range(1, n - 1):
        array.extend([i, i])
    array.extend([n, n, n - 1, n - 1])
    array.extend(list(range(1, n + 1)) + list(range(1, n + 1)))
    print(*array)


def solution():
    t = int(input())
    for _ in range(t):
        solve_testcase()


solution()

# input
# 1
# 2
```

## C. Cost of a Bracket Sequence (01:41)

`-` 삽집을 좀 오래했다

`-` 가장 긴 정규 괄호 부분 문자열의 길이를 구하는 건 쉽다

`-` 왼쪽 괄호는 포함 안 시킬 이유가 없다 (얘는 즉각적으로 길이가 증가하는 건 아니지만 추후 오른쪽 문자열이 포함될 시 도움이 된다)

`-` 오른쪽 괄호는 포함 시킬 수 있다면 무조건 포함 시켜야 한다. 포함 시켜야 길이가 증가한다

`-` 전체 문자열에서 가장 긴 정규 괄호 부분 문자열을 없애보자. 그럼 남은 문자열은 쓸모가 없는 것이며 이를 제거해도 비용이 줄지 않는다

`-` 이때 남은 것의 형태는 `( ... (` 또는 `) ... )` 또는 `) ... ) ( ... (` 또는 빈 문자열이다

`-` 이 외에는 존재할 수 없다. 만약 있다면 올바른 괄호 문자열을 만들 수 있으니 제거할 수 있기 때문이다

`-` 빈 문자열이라면 왼쪽 괄호만 제거하자

`-` 1번 또는 2번인 경우 $k$개는 모두 반대편 괄호로 하면 된다

`-` 3번의 경우 마지막 `)` 앞에 있는 `(`를 모두 없애자. 남으면 첫 `(` 뒤에 있는 `)`를 모두 없애자

`-` 스택을 사용하면 간단하게 구현할 수 있다 (스택에 인덱스 저장)

`-` 시간 복잡도는 $O(n)$이다. $n$이 최대 $5000$이니까 $O\left(n^2\right)$ 풀이를 의도했을텐데 뭘까?

```python
def remove_longest_regular_bracket_sequence(s):
    stack = []
    for i, s_i in enumerate(s):
        if not stack:
            stack.append(i)
            continue
        if s_i =="(":
            stack.append(i)
            continue
        if s[stack[-1]] == "(":
            stack.pop()
        else:
            stack.append(i)
    return stack


def solve_testcase():
    n, k = map(int, input().split())
    s = input().rstrip()
    stack = remove_longest_regular_bracket_sequence(s)
    answer = []
    target = None
    if not stack or (s[stack[0]] == s[stack[-1]] == ")"):
        target = "("
    elif s[stack[0]] == s[stack[-1]] == "(":
        target = ")"
    if target is not None:
        for s_i in s:
            if s_i == target and k > 0:
                answer.append(1)
                k -= 1
            else:
                answer.append(0)
        return answer
    turn = 0
    for i in stack:
        if s[i] == "(":
            turn = i
            break
    for i, s_i in enumerate(s):
        if s_i == "(" and k > 0 and i < turn:
            answer.append(1)
            k -= 1
        elif s_i == ")" and k > 0 and i > turn:
            answer.append(1)
            k -= 1
        else:
            answer.append(0)
    return answer


def solution():
    t = int(input())
    for _ in range(t):
        answer = solve_testcase()
        print("".join(map(str, answer)))


solution()

# input
# 1
# 2 1
# ))
```

## D. Goods on the Shelf (Upsolving)

`-` 업솔빙 했는데 핵 당했다...

`-` 다시 느끼는 거지만 딕셔너리는 매우 느리다. 최대한 배열을 쓰도록 노력하자

`-` 이 문제의 경우 좌표 압축을 수행하면 딕셔너리 대신 배열로 중복 체크를 할 수 있다

`-` 문제를 제대로 풀어보자

`-` 만약 어떤 $x$에 대해 $x$가 등장하는 개별 그룹이 $4$개 이상이면 정렬 불가능이다

`-` 그룹이 $3$개라면 각각이 $1$칸씩만 떨어져있고 적어도 그룹 하나의 원소 개수는 $1$이어야 한다. 그럼 해당 원소를 두 그룹 사이에 집어 넣어서 연속되게 만들 수 있다

`-` 그룹이 $2$개라고 하자. 이 경우 그룹의 원소 개수가 $1$인지 아닌지가 중요하다

`-` 만약 $1$이라면 해당 원소를 나머지 그룹의 양 끝 중 한 곳으로 옮기면 된다 (제미나이가 찾아줬다)

`-` 그렇지 않다면 우선 $1$칸씩만 떨어져 있어야 하며 두 그룹 사이로 원소를 옮겨야 한다

```python
def compress(array):
    sorted_unique = sorted(set(array))
    rank_dict = {v: i for i, v in enumerate(sorted_unique)}
    compressed = [rank_dict[a] for a in array]
    return compressed


def decompose(array):
    n = len(array)
    groups = [[] for _ in range(n + 1)]
    for i, a in enumerate(array):
        if not groups[a]:
            groups[a].append((i, i))
            continue
        s, e = groups[a][-1]
        if e + 1 == i:
            groups[a].pop()
            groups[a].append((s, i))
        else:
            groups[a].append((i, i))
    return groups


def check_two(array, group):
    n = len(array)
    (s1, e1), (s2, e2) = group
    if e1 - s1 >= 1 and e2 - s2 >= 1 and e1 + 2 != s2:
        return "NO"
    if s1 == e1 and e2 + 1 < n and check_valid_swap(array, s1, e2 + 1):
        return "YES"
    if s2 == e2 and s1 - 1 >= 0 and check_valid_swap(array, e2, s1 - 1):
        return "YES"
    if check_valid_swap(array, s1, s2 - 1):
        return "YES"
    if check_valid_swap(array, e2, e1 + 1):
        return "YES"
    return "NO"


def check_three(array, group):
    (s1, e1), (s2, e2), (s3, e3) = group
    if s1 == e1 and e2 + 2 == s3 and check_valid_swap(array, s1, e2 + 1):
        return "YES"
    if s3 == e3 and e1 + 2 == s2 and check_valid_swap(array, e3, e1 + 1):
        return "YES"
    return "NO"


def check_valid_swap(array, i, j):
    array[i], array[j] = array[j], array[i]
    if check_continuous(array):
        array[i], array[j] = array[j], array[i]
        return True
    array[i], array[j] = array[j], array[i]
    return False


def check_continuous(array):
    global VISITED
    VISITED += 1
    seen = SEEN
    visited = VISITED
    n = len(array)
    last = array[0]
    for a in array:
        if seen[a] == visited and last != a:
            return False
        seen[a] = visited
        last = a
    return True


def solve_testcase():
    global VISITED, SEEN
    n = int(input())
    array = list(map(int, input().split()))
    array = compress(array)
    groups = decompose(array)
    VISITED = 0
    SEEN = [0] * (n + 1)
    for group in groups:
        k = len(group)
        if k <= 1:
            continue
        if k > 3:
            print("NO")
            return
        if k == 2:
            print(check_two(array, group))
            return
        if k == 3:
            print(check_three(array, group))
            return
    print("YES")


def solution():
    t = int(input())
    for _ in range(t):
        solve_testcase()


solution()

# input
# 1
# 2
# 5 5
```

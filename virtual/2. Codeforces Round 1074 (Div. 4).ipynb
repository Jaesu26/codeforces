## A. Perfect Root (00:01)

`-` 모든 자연수는 perfect root이다

`-` 따라서 $1$부터 $n$까지 출력하면 된다

```python
def solve_testcase(x):
    return list(range(1, x + 1))


def solution():
    t = int(input())
    for _ in range(t):
        n = int(input())
        answer = solve_testcase(n)
        print(*answer)


solution()

# input
# 1
# 5
```

## B. Prefix Max (00:08)

`-` $a$의 최댓값을 맨 앞에 배치해야 배열의 값을 가장 크게 만들 수 있다

`-` 그럼 정답은 $n\times \max(a)$가 된다

`-` $n$을 입력받는 걸 빼먹었는데 테스트를 안 하고 제출해서 $1$번 틀렸다. 그래서 다시 제출하느라 정답 맞히는데 조금 오래 걸림...

```python
def solve_testcase(array):
    return max(array) * len(array)


def solution():
    t = int(input())
    for _ in range(t):
        n = int(input())
        array = list(map(int, input().split()))
        answer = solve_testcase(array)
        print(answer)


solution()

# input
# 1
# 1
# 1
```

## C. Shifted MEX (00:11)

`-` $a$를 오름차순으로 정렬하자. 이때 $a$의 가장 긴 연속 부분 수열을 $v, v+1, \dots, v+ k$이라 하자

`-` $x=v$로 설정하면 $\operatorname{MEX}(a) = k+1$이 된다

`-` 중복 원소 처리에 유의하자

```python
def solve_testcase(array):
    array.sort()
    max_length = 0
    last = array[0] - 1
    length = 0
    for a in array:
        if a == last + 1:
            length += 1
        else:
            length = 1
        last = a
        max_length = max(length, max_length)
    return max_length


def solution():
    t = int(input())
    for _ in range(t):
        n = int(input())
        array = list(set(map(int, input().split())))
        answer = solve_testcase(array)
        print(answer)


solution()

# input
# 1
# 1
# 1
```

## D. OutOfMemoryError (00:59)

`-` 해시로 안 뚫려서 $25$분을 더 사용했다. 파이썬 해시 저격 테케가 있는 걸까?

`-` 가장 최근에 리셋이 발생한 차례를 기억하자. 그리고 각 $i$별로 리셋이 발생해서 $a_i$를 원래의 값으로 초기화한 차례를 저장해놓자

`-` 만약 이것이 리셋 발생 차례보다 먼저라면 아직 $a_i$를 초기화하지 않았다는 뜻이니 초기화를 진행하면 된다

`-` 전체 알고리즘의 시간 복잡도는 $O(n + m)$이다

```python
def solve_testcase(array, queries, h):
    n = len(array)
    original = array[:]
    last_reset = -1
    last_updated = [-1] * n 
    for i, (b, c) in enumerate(queries):
        b -= 1
        if last_updated[b] < last_reset:
            array[b] = original[b]
            last_updated[b] = i
        array[b] += c
        if array[b] > h:
            array[b] = original[b]
            last_reset = i
            last_updated[b] = i
    for b in range(n):
        if last_updated[b] < last_reset:
            array[b] = original[b]
    return array


def solution():
    t = int(input())
    for _ in range(t):
        n, m, h = map(int, input().split())
        array = list(map(int, input().split()))
        queries = [list(map(int, input().split())) for _ in range(m)]
        answer = solve_testcase(array, queries, h)
        print(*answer)


solution()

# input
# 1
# 1 1 1
# 1
# 1 1
```

## E. The Robotic Rush (01:55)

`-` 영어 이슈로 문제 잘못 해석해서 30분 날리고 중간에 서버 팅겨서 15분을 또 날렸다...

`-` 쿼리마다 모든 로봇이 해당 방향으로 움직여야 한다. 각 로봇별로 스파이크에 닿지 않고 왼쪽, 오른쪽으로 몇 칸 씩 이동할 수 있는지 이분 탐색을 이용해 계산하자

`-` 왼쪽 여유분을 기준으로 내림차순 정렬한 배열과 오른쪽 여유분을 기준으로 내림차순 정렬한 배열, 그리고 각 로봇이 스파이크에 닿아 죽었는지 여부를 저장하는 배열을 만들자

`-` 누적된 쿼리에 의해 현재 왼쪽 또는 오른쪽으로 몇 칸 움직여야 되는지를 추적하자. 만약 왼쪽으로 $l$칸 움직여야 하고 가장 작은 왼쪽 여유분이 $l$보다 작다면 해당 로봇은 스파이크에 닿아 죽는다

`-` 따라서 죽었다고 표시를 해두며 배열에서 제거하자. 모든 로봇은 한 번 제거되면 다시 추가되지 않으며 최소 여유분만 확인하면 되므로 $O(n)$이다

`-` 전체 알고리즘의 시간 복잡도는 $O(n\log n + m\log m + k)$이다

```python
from bisect import bisect_left


def solve_testcase(A, B, instructions):
    inf = 10**9
    n, m = len(A), len(B)
    left_spare, right_spare = [0] * n, [0] * n
    is_dead = [False] * n
    num_survivors = n
    for i, a in enumerate(A):
        j = bisect_left(B, a)
        left_spare[i] = A[i] - B[j - 1] if j != 0 else inf
        right_spare[i] = B[j] - A[i] if j != m else inf
    left_order = sorted(range(n), key=lambda i: -left_spare[i])
    right_order = sorted(range(n), key=lambda i: -right_spare[i])
    left_move = right_move = 0
    answer = []
    for s in instructions:
        if s == "L":
            left_move += 1
            right_move -= 1
            while left_order and left_spare[left_order[-1]] <= left_move:
                i = left_order.pop()
                if is_dead[i]:
                    continue
                is_dead[i] = True
                num_survivors -= 1
            answer.append(num_survivors)
            continue
        left_move -= 1
        right_move += 1
        while right_order and right_spare[right_order[-1]] <= right_move:
            i = right_order.pop()
            if is_dead[i]:
                continue
            is_dead[i] = True
            num_survivors -= 1
        answer.append(num_survivors)
    return answer


def solution():
    t = int(input())
    for _ in range(t):
        n, m, k = map(int, input().split())
        A = list(map(int, input().split()))
        B = list(map(int, input().split()))
        A.sort()
        B.sort()
        instructions = input().rstrip()
        answer = solve_testcase(A, B, instructions)
        print(*answer)


solution()

# input
# 1
# 1 1 1
# 1
# 2
# L
```

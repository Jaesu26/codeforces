## A. Square? (00:01)

`-` $a,b, c, d$가 모두 같은지 확인하자

```python
def solve_testcase(a, b, c, d):
    if a == b == c == d:
        return "YES"
    return "NO"


def solution():
    t = int(input())
    for _ in range(t):
        a, b, c, d = map(int, input().split())
        answer = solve_testcase(a, b, c, d)
        print(answer)


solution()

# input
# 1
# 1 1 1 1
```

## B. Your Name (00:05)

`-` $s$와 $t$를 정렬한 뒤 둘이 동일한지 판단하면 된다

```python
def solve_testcase(s, t):
    ss = "".join(sorted(list(s)))
    tt = "".join(sorted(list(t)))
    if ss == tt:
        return "YES"
    return "NO"


def solution():
    q = int(input())
    for _ in range(q):
        n = int(input())
        s, t = input().rstrip().split()
        answer = solve_testcase(s, t)
        print(answer)


solution()

# input
# 1
# 1
# a a
```

## C. Isamatdin and His Magic Wand! (00:15)

`-` **lexicographically**는 **사전순으로**라는 의미이다 (이걸 몰랐어..., 이젠 앎)

`-` 정의가 나와있긴 한데 정확히 해석하지 못했다. 정의가 조금 애매한 것 같다. $j$가 존재하지 않으면 어떻게 되는가? 근데 그래도 의미를 파악할 순 있었다. 나는 그냥 이상하게 해석했을뿐

`-` 아무튼 사전순으로 가장 작게 만들면 된다

`-` $x,y$와 $t$의 홀짝성이 다르다고 해보자. 그럼 $x$와 $t$를 교환하고 $y$와 $t$를 교환한 뒤 다시 $x$와 $t$를 교환하면 결과적으로 $x$와 $y$만 교환한 결과가 된다

`-` 따라서 정수가 전부 홀수 또는 전부 짝수만 아니라면 원하는대로 배치할 수 있으니 오름차순으로 정렬을 수행하자. 그게 아니라면 아무 장난감도 교환할 수 없으니 입력된 $n$개의 정수를 그대로 출력하면 된다

```python
def solve_testcase(array):
    r1 = all(a % 2 == 0 for a in array)
    r2 = all(a % 2 == 1 for a in array)
    if r1 or r2:
        return array
    return sorted(array)


def solution():
    t = int(input())
    for _ in range(t):
        n = int(input())
        a = list(map(int, input().split()))
        answer = solve_testcase(a)
        print(*answer)


solution()

# input
# 1
# 1
# 1
```

## D. Yet Another Array Problem (00:33)

`-` 배열 내 원소와 서로소인 $2$ 이상의 정수 중 가장 작은 수를 출력하면 된다

`-` $a_i$는 $10^{18}$ 이하이다.  $2^{60} > 10^{18}$이므로 $a_i = p_1 \times p_2 \times \cdots \times p_k$라고 할 때 $k$는 $60$을 넘지 못한다 ($p$는 소수)

`-` $1000$ 이하의 소수 개수는 $60$ 이상임을 참고하자. 에라토스테네스의 체를 이용해 $2$ 이상 $1000$ 이하의 소수를 찾은 뒤 작은순으로 $60$개만 취해서 가지고 있자. $1$번만 계산해두면 전체 테스트 케이스에서 사용할 수 있다

`-` 가장 작은 소수 $2$부터 시작해 배열 내 원소와 $\gcd$를 계산하자. 만약 $1$을 도출하는 원소가 존재한다면 해당 소수가 정답이다. 그렇지 않다면 다음 소수로 넘어가서 탐색을 재개하면 된다

```python
import math


def sieve_of_eratosthenes(n):
    is_primes = [False] * 2 + [True] * (n - 1)
    for i in range(2, n + 1):
        if i * i > n:
            break
        if not is_primes[i]:
            continue
        for j in range(i * i, n + 1, i):
            is_primes[j] = False
    prime_nums = [i for i, is_prime in enumerate(is_primes) if is_prime]
    return prime_nums


def solve_testcase(array, prime_nums):
    for p in prime_nums:
        for a in array:
            g = math.gcd(p, a)
            if g == 1:
                return p
    return -1


def solution():
    prime_nums = sieve_of_eratosthenes(1000)
    prime_nums = prime_nums[:60]
    t = int(input())
    for _ in range(t):
        n = int(input())
        a = list(map(int, input().split()))
        answer = solve_testcase(a, prime_nums)
        print(answer)


solution()

# input
# 1
# 1
# 1
```

## F. Tree, TREE!!! (02:04)

`-` 루트 노드를 $r$이라 하자. 임의의 노드 $u$를 LCA 아웃풋으로 만들고 싶다. 이는 단순히 $u$의 서브 트리 크기가 $k$ 이상이면 된다

`-` 왜냐하면 $u$를 고른 뒤 $u$의 서브 트리에서 $k-1$개의 노드를 종류에 상관없이 고르면 이들의 LCA가 $u$가 되기 때문이다

`-` 따라서 서브 트리 크기가 $k$ 이상인 노드의 개수가 $|S_r|$이 된다

`-` 각 노드별 서브 트리 크기는 $O(n)$에 계산할 수 있다. 근데 각 노드를 루트로 가정한 뒤 서브 트리 크기를 매번 새로 계산하는 건 $O\left(n^2\right)$이라 시간 초과이다

`-` $r$의 자식 노드 $c_1,c_2,\dots, c_i$를 고려하자. 만약 $c_1$이 새로운 루트가 되면 기존 서브 트리 크기와 비교하여 달라지는 건 $r$과 $c_1$뿐이다. $c_1$의 서브 트리 크기는 당연하게도 $n$이 된다. $r$의 서브 트리 크기는 기존 $n$에서 기존 $c_1$의 서브 트리 크기를 차감한 것이 된다. $c_1$의 서브 트리 크기를 $x$라 할 때 기존 $n, x$에서 $n - x, n$으로 변하게 된다

`-` 따라서 $|S_{c_1}|= |S_r| - \mathbf{1}_{[k, \infty)}(x) + \mathbf{1}_{[1, n - k]}(x)$가 된다 (단, $\mathbf{1}$는 지시 함수)

`-` 처음에 각 노드별 서브 트리와 $|S_r|$을 계산해두면 그 이후에는 $|S_{c_i}|$를 $O(1)$에 계산할 수 있다. 이를 재귀적으로 반복하면 되므로 $\sum\limits_{r=1}^{n}|S_r|$을 $O(n)$에 계산할 수 있다

```python
from collections import deque


def dfs(tree, root):
    n = len(tree)
    parents = [0] * n
    order = []
    stack = [root]
    while stack:
        node = stack.pop()
        order.append(node)
        for child in tree[node]:
            if child == parents[node]:
                continue
            parents[child] = node
            stack.append(child)
    return parents, order


def compute_subtree_sizes(tree, parents, order):
    subtree_sizes = [1] * len(tree)
    for node in reversed(order):
        subtree_sizes[parents[node]] += subtree_sizes[node]
    return subtree_sizes


def bfs(tree, root, subtree_sizes, count, k):
    n = len(tree) - 1
    queue = deque([(root, 0, count)])
    total = 0
    while queue:
        node, parent, count = queue.popleft()
        total += count
        for child in tree[node]:
            if child == parent:
                continue
            x = subtree_sizes[child]
            queue.append((child, node, count - (x >= k) + (x <= n - k)))
    return total


def solve_testcase(tree, root, k):
    parents, order = dfs(tree, root)
    subtree_sizes = compute_subtree_sizes(tree, parents, order)
    count = sum(1 for u in range(1, len(tree)) if subtree_sizes[u] >= k)
    total = bfs(tree, root, subtree_sizes, count, k)
    return total


def solution():
    root = 1
    t = int(input())
    for _ in range(t):
        n, k = map(int, input().split())
        tree = [[] for _ in range(n + 1)]
        for _ in range(n - 1):
            u, v = map(int, input().split())
            tree[u].append(v)
            tree[v].append(u)
        answer = solve_testcase(tree, root, k)
        print(answer)


solution()

# input
# 1
# 2 2
# 1 2
```

## G. Mukhammadali and the Smooth Array (01:26)

`-` 마지막 문제인데 이게 쉬운지 E, F보다 G가 더 많이 풀렸다

`-` 이거 예전에 본 문제다 ㄷㄷ. 풀진 않고 보기만 했는데 대충 어떤 식으로 접근하는지도 봤다. 근데 봤다는 기억만 나고 내용이 기억나지 않는다

`-` 모든 인덱스를 *drop* 하지 않기 위해 배열 내 원소를 교체해야 한다. 이때 교체 비용을 최소로 하면 된다

`-` 교체 후 배열은 비내림차순으로 정렬되어 있다. 이 말은 배열이 비내림차순이 되도록 원소를 적절히 교체해 나가야 한다는 것이다

`-` $i$번째 원소를 교체하거나 교체하지 않을 수 있는데 교체하지 않는다고 해보자. 이 뜻은 $m_i = \max(a'_1, a'_2, \dots, a'_{i-1}) \le a_i$를 만족한다는 것이다 ($a'_j$는 $j$번째 원소의 교체 고려를 끝낸 뒤의 정수를 의미, 교체했거나 교체하지 않았음)

`-` 만약 교체한다고 해보자. 그럼 원하는 정수로 교체할 수 있다. 아무 원소로 교체해도 상관없지만 비내림차순을 유지하기만 하면 되는 것에 주목하자. $a_i$를 $m_i$로 교체할 것이다. 그럼 $m_i$는 $0, a_1, a_2,\dots, a_{i-1}$ 중 하나에 바인딩 된다 ($0$은 모든 원소를 교체한 경우)

`-` 만약 아무렇게나 원소를 원하는 정수로 교체했다면 고려해야 될 경우의 수가 매우 많았을 것이다. 하지만 특정 원소에만 바인딩 시켰으므로 $i$번째 원소를 교체할 때 고려할 경우는 $i$개 뿐이다

`-` 계산 편의성을 위해 맨 앞에 $0$을 삽입한 배열을 대신 고려하자

`-` $\operatorname{dp}[x]$를 여태까지 교체 고려를 끝낸 원소들 중 최댓값이 $a_x$일 때의 최소 교체 비용이라 하자. 처음 $\operatorname{dp}$ 배열의 원소는 $\infty$로 초기화 되었으며 $\operatorname{dp}[0] = 0$이다

`-` 현재 $i$번째 원소의 교체를 고려하고 있으며 이전 원소들이 $a_j$에 바운드 됐다고 하자 ($j < i$). 만약 $a_i \ge a_j$라면 $i$번째 원소를 교체하지 않아도 되며 최댓값은 $a_i$로 바뀐다. 따라서 $\operatorname{dp}[i] = \min(\operatorname{dp}[j], \operatorname{dp}[i])$이다. 그렇지 않거나 $a_i \ge a_j$라고 해도 미래를 위해 $a_i$를 교체해야 될 수 있다. 따라서 $c_i$의 비용을 사용해 $a_i$를 $a_j$로 교체하자. 이러면 $\operatorname{dp}[j] = \operatorname{dp}[j] + c_i$가 된다. $i$는 $1$부터 $n$까지 이며 순회를 마친 뒤의 $\min(\operatorname{dp})$가 정답이 된다

`-` 전체 알고리즘의 시간 복잡도는 $O\left(n^2\right)$이며 토글링을 사용해서 공간 복잡도는 $O(n)$이다. $n$은 최대 $8000$이므로 TLE, MLE 걱정은 없다

```python
def solve_testcase(a, c):
    n = len(a)
    inf = float("inf")
    dp = [inf] * n
    dp[0] = 0
    for i in range(1, n):
        for j in range(i):
            if a[i] >= a[j]:
                dp[i] = min(dp[j], dp[i])
            dp[j] += c[i]
    return min(dp)


def solution():
    t = int(input())
    for _ in range(t):
        n = int(input())
        a, c = [0], [0]
        a.extend(map(int, input().split()))
        c.extend(map(int, input().split()))
        answer = solve_testcase(a, c)
        print(answer)


solution()

# input
# 1
# 1
# 1
# 1
```

## A. Nene's Game (00:09)

`-` 현재 플레이어 $n$명이 있다고 하자. $a$의 원소 중 $x$개가 $n$보다 작거나 같으면 $x$명이 퇴출되므로 $n$을 $n-x$로 갱신하자. $n$이 $a_1$보다 작아질 때까지 시뮬레이션을 진행하면 된다

```python
def solve_testcase(array, queries):
    q = len(queries)
    answers = [0] * q
    for i, n in enumerate(queries):
        n_people = n
        while n_people >= array[0]:
            x = sum(1 for a in array if a <= n_people)
            n_people -= x
        answers[i] = n_people
    return answers


def solution():
    t = int(input())
    for _ in range(t):
        k, q = map(int, input().split())
        a = list(map(int, input().split()))
        n = list(map(int, input().split()))
        answers = solve_testcase(a, n)
        print(*answers)


solution()
```

`-` 에디토리얼 보니까 쉬운 풀이가 존재한다. 시뮬레이션을 반복하다 보면 결국에 플레이어 수는 $a_1 - 1$까지 줄어들게 된다. 따라서 $\min(n, a_1 - 1)$이 정답이다

`-` 나는 그냥 지문 읽고 $t,k,q$ 범위가 작아 시뮬레이션으로 풀리니까 해당 접근을 바로 구현했다. 여기에 매몰되어서 다른 생각이 떠오르지 않았다

## B. Nene and the Card Game (00:19)

`-` 내가 가지고 있는 카드 중 중복된 카드를 먼저 내려놓아야 한다. 그렇지 않으면 나머지 한 장을 Nene가 가지고 있을테니 Nene만 $1$점을 얻게 된다

`-` 내가 소유하고 있는 중복된 카드의 종류 수를 $x$라 하고 Nene는 $y$라 하자. 만약 $x \le y$라면 나는 $x$점밖에 획득하지 못한다. 그렇지 않다면 나는 $n - y$점을 획득할 수 있다 

```python
def solve_testcase(array):
    n = len(array)
    n_unique = set(map(str, array))
    x = n - len(n_unique)
    y = n - len(n_unique)
    if x <= y:
        return x
    return n - y


def solution():
    t = int(input())
    for _ in range(t):
        n = int(input())
        a = list(map(int, input().split()))
        answer = solve_testcase(a)
        print(answer)


solution()
```

`-` 에디토리얼 보니까 $x=y$라고 한다. 웃긴 게 내 코드를 보니까 $x$랑 $y$에 동일한 값을 할당시켜 놨다

`-` 이게 같은 값이지만 의미하는 바는 다르다. $a$의 고유한 수의 개수를 $k$라 하자. 내가 받은 카드 $n$장 중 중복 카드가 없다면 $k=n$일 것이다. 나는 카드를 $n$장 가지고 있는데 고유한 카드가 $k$장이라는 건 $n-k$장은 내가 중복으로 가지고 있다는 의미이다. 따라서 $x=n-k$이다

`-` 한편, Nene는 내가 가지고 있지 않은 카드를 소유하고 있다. 따라서 내가 $1$장도 가지고 있지 않은 카드를 중복해서 가지고 있을 것이다. 그러므로 $y=n-k$이다

`-` 구현은 나의 논리를 파이썬으로 번역하는 것이다 보니 의미가 다른 두 변수가 같은 값을 가지리라 알아차리지 못했다

## C. Nene's Magical Matrix (01:58)

`-` 수많은 시도끝에 맞힌 문제이다. $n=1,2$인 테케밖에 없어서 왜 틀렸는지 알기 힘들었다

- 첫 번째 시도

`-` $p = \{1,2,\dots, n\}$이라 하자. 열마다 $p$로 채우자. 그리고 행의 합이 $\frac{n(n+1)}{2}$ 미만이라면 $p$로 대체하자

`-` 정답이 아니길래 쿼리를 잘못 저장했나 생각했지만 계속 틀리길래 더 큰 $s$가 존재하는 걸로 생각했다

`-` $n=3$인 경우 도출한 최종 매트릭스는 다음과 같다

$$\begin{bmatrix}
1 & 2 & 3  \\
2 & 2 & 2  \\
3 & 3 & 3  \\
\end{bmatrix}$$

- 두 번째 시도

`-` 열마다 $p$로 채운 뒤 행의 합이 $\frac{n(n+1)}{2}$ 이하라면 $p$로 대체하자. 그리고 다시 열을 봤을 때 합이 $\frac{n(n+1)}{2}$ 이하라면 $p$로 대체하자. 단, 연산은 $2n$번까지만 적용하자

`-` 정답이 아니다. 더 큰 $s$를 찾아보자

`-` $n=3$인 경우 도출한 최종 매트릭스는 다음과 같다

$$\begin{bmatrix}
1 & 2 & 3  \\
2 & 2 & 3  \\
3 & 3 & 3  \\
\end{bmatrix}$$

- 세 번째 시도

`-` 두 번째 시도의 방법을 확장시키자. 행과 열을 번갈아가며 합이 $\frac{n(n+1)}{2}$ 이하라면 $p$로 대체하자. 단, 연산은 $2n$번까지만 적용하자

`-` 이러면 정답일 줄 알았지만 정답이 아니다. 쿼리가 잘못된 걸까? 설마 더 큰 $s$가 존재하나?

`-` $n=4$인 경우 최종 매트릭스는 다음과 같다

$$\begin{bmatrix}
1 & 2 & 3 & 4  \\
2 & 2 & 3 & 4 \\
3 & 3 & 3 & 3 \\
4 & 4 & 4 & 4 \\
\end{bmatrix}$$

- 네 번째 시도

`-` 혹시 정답 매트릭스는 이런 꼴이 아닐까?

$$\begin{bmatrix}
1 & 2 & 3 & \dots & n \\
2 & 2 & 3 & \dots & n \\
3 & 3 & 3 & \dots & n \\
\vdots & \vdots & \vdots & \ddots & \vdots \\
n & n & n & \dots & n
\end{bmatrix}$$

`-` $n=3$인 경우는 만족한다. 근데 $n=4$인 경우 $3$행 $4$열의 값이 $3$이다. 이걸 $4$로 바꿀 수 있나? 나는 합과 열에 대해 합이 $\frac{n(n+1)}{2}$ 이하인 경우에만 $p$로 대체했다. 하지만 이것과 상관없이 정답 매트릭스에 맞추도록 조작해봤다. 그랬더니 정답 매트릭스를 만들 수 있었다!

`-` 문제는 연산 횟수가 $2n$을 넘으면 안 된다는 것이다. 매트릭스에 $n$만 $2n-1$개 존재해서 $2n-1$번의 연산은 $n$을 배치하는 데 사용되며 나머지 $1$번의 연산으로 정답 매트릭스를 만들어야 한다. 이게 가능한가?

`-` 그렇다고 이보다 합이 더 크거나 같은 다른 꼴의 매트릭스가 존재할 것 같지도 않다. $n$행과 $n$열에 $n$을 배치하면서도 정답 매트릭스와 유사한 형태가 되도록 최적의 연산을 진행해야 된다

`-` 이미 정답인 배치를 덮어쓰는 경우를 줄이기 위해 행과 열을 번갈아가며 연산을 진행해야 할 것 같았고 그 결과 $n$행 $\to$ $n-1$열 $\to \dots \to$ $1$행순으로 $p$를 할당하면 정답 매트릭스를 $2n-1$번의 연산에 유도할 수 있게 된다 ($s = \sum\limits_{i=1}^{n} i \cdot (2i - 1)$이다)

```python
def solve_testcase(n):
    p = list(range(1, n + 1))
    queries = []
    row, col = n, n - 1
    while col > 0:
        queries.append([1, row] + p)
        queries.append([2, col] + p)
        row -= 1
        col -= 1
    queries.append([1, 1] + p)
    s = sum([i * (2 * i - 1) for i in range(1, n + 1)])
    return s, len(queries), queries


def solution():
    t = int(input())
    for _ in range(t):
        n = int(input())
        s, m, queries = solve_testcase(n)
        print(s, m)
        for q in queries:
            print(*q)


solution()

# input
# 1
# 3
```

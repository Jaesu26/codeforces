## B. Skibidus and Ohio (00:04)

`-` 대체하고 삭제하나 삭제하고 대체하나 결과는 동일하다. 따라서 가능한 모든 인접한 두 문자에 대해 동일한 것이 있다면 $1$, 아니라면 $|s|$가 정답이다 (삭제 후 갱신된 인접 문자로 바꿔나가면 계속 삭제 가능)

```python
def solve_testcase(string):
    prev = None
    for s in string:
        if s == prev:
            return 1
        prev = s
    return len(string)


def solution():
    t = int(input())
    for _ in range(t):
        s = input().rstrip()
        answer = solve_testcase(s)
        print(answer)


solution()

# input
# 4
# baa
# skibidus
# cc
# ohio
```

## C2. Skibidus and Fanum Tax (hard version) (00:23)

`-` C1번 문제 풀이는 C2번 문제 풀이로 갈음한다

`-` 연산을 적절히 적용해서 배열 $a$를 비내림차순으로 만들 수 있는지 확인하는 문제이다. $a$를 순서대로 순회하며 연산을 적용할지 말지 정할 것이다

`-` 우선 $a_1$은 작을수록 비내림차순으로 만드는 데 유리하므로 배열 $b$의 최솟값을 골라 $a_1$을 $\min(a_1, \min\limits_{1 \le j \le m} \{b_j\} - a_1)$으로 변경하자. 이제 $i>1$인 $a_i$에 대해 $a_{i-1}$보다 큰 값 중 가장 작은 값을 가지도록 연산을 적용할 것이다

`-` 만약 연산을 적용하지 않는다면 $a_i$이다. 연산을 적용하면 $b_j - a_i$인데 여기서 $b_j$는 배열 $b$에서 $a_{i-1} + a_i$보다 큰 수 중 가장 작은 수이며 이를 빠르게 찾기 위해 이분 탐색을 수행하자 ($b$를 정렬해두자). $a_i$와 $b_j - a_i$ 중 적절한 값을 선택하면 되며 만약 둘 다 $a_{i-1}$보다 작다면 정답은 NO이다

`-` 전체 알고리즘의 시간 복잡도는 $O((n + m)\log m)$이다. 구현의 편의상 $\min\limits_{1 \le j \le m} \{b_j\}$도 이분 탐색으로 찾았다 ($a_0 = -\infty$)

```python
from bisect import bisect_left


def solve_testcase(a, b):
    m = len(b)
    b.sort()
    prev = -float("inf")
    for a_i in a:
        j = bisect_left(b, prev + a_i)
        b_j = b[j] if j < m else b[-1]
        c_i = b_j - a_i
        if c_i < prev and a_i < prev:
            return "NO"
        if c_i > a_i:
            a_i, c_i = c_i, a_i
        if prev <= c_i:
            prev = c_i
        else:
            prev = a_i
    return "YES"


def solution():
    t = int(input())
    for _ in range(t):
        n, m = map(int, input().split())
        a = list(map(int, input().split()))
        b = list(map(int, input().split()))
        answer = solve_testcase(a, b)
        print(answer)


solution()

# input
# 1
# 3 2
# 1 4 3
# 3 4
```

## D. Skibidus and Sigma (00:46)

`-` 문제 이해가 잘 안돼서 오래 걸렸다. 배열 내 원소의 순서는 변경할 수 없고 배열 간의 순서만 조정할 수 있다 

`-` 이 문제는 직관으로 풀었다. 딱 보면 그리디임 (이러면 안 돼요)

`-` 배열의 합계를 기준으로 내림차순 정렬한 순서가 최적이며 이는 exchange argument에 의해 성립한다. 증명은 다음과 같다

`-` 임의의 배열 $c$에 대해 $c$의 합계를 $T_c$라 하자. 그리고 $c$의 score를 $F_c$라 표현하자. $n$개의 배열을 연결해서 만든 배열의 score는 $F^*$로 표현할 것이다

`-` 두 배열 $a,b$가 있을 때 $a,b$ 순서로 연결한 배열을 $ab$라 하면 $F_{ab} =  2m \cdot a_1 + (2m-1) \cdot a_2 + \cdots + (m + 1) \cdot a_m + m \cdot b_1 + (m-1) \cdot b_2 + \cdots + 1 \cdot b_m = mT_a + F_a + F_b$이다

`-` 반대로 $F_{ba} = mT_b + F_a + F_b$이다. $a,b$ 순서로 합치든 $b,a$ 순서로 합치든 나머지 배열의 위치는 그대로이니 이들이 $F^*$에 끼치는 영향은 똑같음에 주목하자. $T_{ab} > T_{ba}$라면 $F_{ab} > F_{ba}$이고 반대로 $T_{ab} < T_{ba}$라면 $F_{ab} < F_{ba}$이다. 따라서 합계가 큰 순으로 앞에서부터 배치하는 게 score를 가장 크게 만드는 전략이다

`-` score는 $O(m)$에 계산할 수 있다. 그러므로 전체 알고리즘의 시간 복잡도는 $O(n\log n + nm)$이다

```python
def compute_score(array):
    m = len(array)
    return sum([a_i * (m - i) for i, a_i in enumerate(array)])


def solve_testcase(arrays):
    arrays.sort(key=sum, reverse=True)
    collected = []
    for array in arrays:
        collected.extend(array)
    score = compute_score(collected)
    return score


def solution():
    t = int(input())
    for _ in range(t):
        n, m = map(int, input().split())
        arrays = [list(map(int, input().split())) for _ in range(n)]
        answer = solve_testcase(arrays)
        print(answer)


solution()

# input
# 1
# 2 2
# 4 4
# 6 1
```

## E. Skibidus and Rizz (01:20)

`-` **balance-value**란 이진 문자열에서 $\mathsf{0}$과 $\mathsf{1}$의 개수 차이를 뜻한다. 가능한 모든 부분 문자열의 **balance-value** 최댓값이 $k$가 되도록 이진 문자열 $s$를 구성하면 된다

`-` **balance-value**를 극대화하는 경우는 한 가지 문자열만 사용하는 것이다. 따라서 이론상 부분 문자열의 **balance-value** 최댓값은 $\max(n,m)$이다. 만약 $k > \max(n,m)$이라면 더 볼 것도 없이 불가능이다

`-` 한편, $s$의 **balance-value**는 항상 $|n - m|$이다. 이것이 $k$보다 커도 불가능이다. 그럼 일단 가능한 $k$의 범위는 $|n - m| \le k \le \max(n,m)$이다

`-` $s$의 부분 문자열 중 **balance-value**가 $k$인 문자열 $r$이 존재해야 한다. 이를 고려해보자 ($\mathsf{1}$이 더 많다고 하자). 일반성을 잃지 않고 $n < m$이라 하겠다. $r$의 양 끝 문자는 무조건 $\mathsf{1}$이다. 그렇지 않다면 $\mathsf{1}$을 제거해서 **balance-value**를 $k+1$로 만들 수 있다. 그리고 $r$의 왼쪽, 오른쪽 문자는 $\mathsf{0}$이어야만 한다. 그게 아니라면 $\mathsf{1}$일 테고 이를 포함해서 **balance-value**를 $k+1$로 만들 수 있다

`-` 이를 좀 더 확장해보자. 일반성을 잃지 않고 오른쪽 방향만을 고려하겠다

`-` 오른쪽 방향에 대해 $\mathsf{0}$과 $\mathsf{1}$의 개수 차이를 누적했을 때 이 값이 $0$보다 큰 순간이 있으면 안 된다. 만약 존재하면 해당 부분까지 $r$에 포함시켜서 **balance-value**를 $k$보다 크게 만들 수 있다

`-` 따라서 $r$ 오른쪽에 $\mathsf{0}$과 $\mathsf{1}$을 번갈아가며 배치하고 남은 $\mathsf{0}$을 모두 배치하면 조건을 만족하는 이진 문자열 $s$를 만들 수 있다. 이때 $s$의 오른쪽부터 봤을 때 $n - m + k$ 길이의  $\mathsf{0}$으로만 이루어진 부분 문자열이 존재한다. 이것의 길이가 $k$를 초과하지 않을까 생각할 수 있지만 $n - m \le 0$이므로 문제 없다

```python
def solve_testcase(n, m ,k):
    if k > max(n, m) or k < abs(m - n):
        return -1
    if n < m:
        s = "1" * k + "01" * (m - k) + "0" * (n - m + k)
    else:
        s = "0" * k + "10" * (n - k) + "1" * (m - n + k)
    return s


def solution():
    t = int(input())
    for _ in range(t):
        n, m, k = map(int, input().split())
        answer = solve_testcase(n, m, k)
        print(answer)


solution()

# input
# 2
# 1 2 1
# 4 3 2
```

## F. Skibidus and Slay (Upsolving)

`-` 문제를 이해하는 데 조금 걸렸고 생각지 못한 점이 있어서 TLE를 받았다. 콘테스트가 끝나고 내 코드의 문제점이 뭘까 추론해보니 허점을 찾았고 문제를 해결할 수 있었다

`-` 문제로 돌아와서 $x$가 *majority*인 경로를 생각해보자. 경로의 양 끝 노드의 값은 $x$여야 한다. 그렇지 않다며 해당 노드를 제외해도 $x$가 *majority*인 경로임은 변하지 않으니 이를 반복하면 결국에 양 끝 노드의 값은 $x$가 된다. 즉, 값이 $x$인 노드에서 DFS를 수행해 값이 $x$인 다른 노드에서 탐색을 종료한다는 것이다

`-` $i$번 노드에서 DFS를 수행해보자. 만약 인접 노드의 값이 $a_i$이면 $a_i$는 *majority*이다. 그게 아니라고 해보자. 인접 노드의 인접 노드를 탐색할 것이고 $a_i$를 값으로 가지면 $a_i$는 *majority*이다! ($3$개 중 $2$개가 $a_i$)

`-` 그렇지 않다고 해보자. 그렇다면 $a_i$가 *majority*인 경로가 존재하더라도 $i$번 노드를 포함할 필요가 없다. 왜냐하면 $i$번 노드를 경로에 포함시키기 위해서 $a_i$가 아닌 다른 값을 가지는 불필요한 노드를 $2$개나 포함해야 하기 때문이다. 이는 최빈값 관점에서 봤을 때 손해이다. 따라서 길이가 $3$ 이상인(=노드가 $4$개 이상) 경로는 고려할 가치가 없다

`-` $1$번 노드부터 $n$번 노드까지 시작 노드로 하여 DFS를 수행할 것이지만 길이가 $2$ 이하인 경로만 고려하므로 전체 알고리즘의 시간 복잡도는 $O(n)$이라 생각했다. 하지만 이를 그대로 구현하니 TLE를 받았고 파이썬은 진짜 심각한 언어라 중얼거리며 콘테스트가 끝났다...  

`-` 근데 아무리 파이썬 느려도 TLE가 발생한다는 게 이상해서 좀 더 생각해보니 노드 하나를 중심에 두고 나머지 노드를 중심에 있는 노드와 연결시키면 모든 노드 간의 거리가 $2$ 이하이게 된다. 따라서 어떤 노드를 시작 노드로 하든 DFS를 수행하면 모든 노드를 확인하게 되고 이를 $n$번 수행하니까 총 $O\left(n^2\right)$으로 TLE가 발생하는 게 당연했다 (파이썬이 문제가 아니어서 다행이야)

`-` 참고로 이런 꼴의 그래프를 [성게 그래프](https://en.wikipedia.org/wiki/Star_(graph_theory))라 한다. 예전에 봤었는데 특성을 생각 안 하고 이런 종류의 트리도 있구나만 생각했다. 편향 트리에 더해서 성게 그래프도 내 머릿속에 저장해두자. 호되게 당했다

`-` 다른 방법으로 $u$번 노드를 중심으로 하는 길이가 $2,3$인 경로를 고려하자. 이는 $u$번 노드와 인접한 노드만 확인함으로써 수행할 수 있다. $u$번 노드를 포함해 이들 중 값으로 $x$를 가지는 노드가 $2$개 이상 있다면 $x$는 *majority*이다

`-` 각 노드별로 차수만큼만 노드를 확인하며 모든 노드의 차수 합계는 $2(n-1)$이므로 전체 알고리즘의 시간 복잡도는 $O(n)$이다

```python
def solve_testcase(tree, array, s):
    for u, neighbors in enumerate(tree):
        if not neighbors:
            continue
        a_u = array[u - 1]
        seen = set([str(a_u)])
        for v in neighbors:
            a_v = array[v - 1]
            if s[a_v - 1] == 1:
                continue
            s_v = str(a_v)
            if s_v in seen:
                s[a_v - 1] = 1
                continue
            seen.add(s_v)
    s = "".join(map(str, s))
    return s


def solution():
    t = int(input())
    for _ in range(t):
        n = int(input())
        a = list(map(int, input().split()))
        tree = [[] for _ in range(n + 1)]
        s = [0] * n
        for _ in range(n - 1):
            u, v = map(int, input().split())
            tree[u].append(v)
            tree[v].append(u)
            a_u, a_v = a[u - 1], a[v - 1]
            if a_u == a_v:
                s[a_u - 1] = 1
        answer = solve_testcase(tree, a, s)
        print(answer)


solution()

# input
# 1
# 4
# 2 4 4 2
# 1 2
# 2 3
# 3 4
```

## B. Bobritto Bandito (00:07)

`-` $n-m$만큼 구간이 좁아져야 한다. 시작은 $0$번 집이니까 구간은 항상 $0$을 포함해야 됨에 유의하자 (이것 때문에 1번 틀렸다)

```python
def solve_testcase(n, m, l, r):
    c = n - m
    while l < 0 and c > 0:
        l += 1
        c -= 1
    while r > 0 and c > 0:
        r -= 1
        c -= 1
    return l, r


def solution():
    t = int(input())
    for _ in range(t):
        n, m, l, r = map(int, input().split())
        answer = solve_testcase(n, m, l, r)
        print(*answer)


solution()

# input
# 1
# 4 1 0 4
```

## C. Brr Brrr Patapim (00:15)

`-` 순열은 항상 존재하며 유일하게 존재함을 증명할 수 있다고 한다. 그러니 $i+j \le 2n$을 만족하는 $G_{i,j}$를 순회하며 $p_{i+j}$에 저장하자. $p_1$은 $1$부터 $2n$까지의 수 중 $G_{i,j}(i+j\le2n)$를 순회하며 $1$번도 등장하지 않은 수가 된다

```python
def solve_testcase(G):
    n = len(G)
    m = 2 * n
    p = [0] * m
    seen = [False] * (m + 1)
    for i in range(1, n + 1):
        for j in range(1, n + 1):
            if i + j > m:
                continue
            g = G[i - 1][j - 1]
            p[i + j - 1] = g
            seen[g] = True
    p[0] = next(x for x in range(1, m + 1) if not seen[x])
    return p


def solution():
    t = int(input())
    for _ in range(t):
        n = int(input())
        G = [list(map(int, input().split())) for _ in range(n)]
        answer = solve_testcase(G)
        print(*answer)


solution()

# input
# 1
# 3
# 1 6 2
# 6 2 4
# 2 4 3
```

## D. Tung Tung Sahur (00:36)

`-` 구현을 신속 정확하게 하지 못해 오래 걸렸다

`-` $p$를 순회하며 이전 원소와 달라지면 $s$를 확인하자. 예컨대 여태까지 $L$이 $c$번 등장했고 현재 $p_i$가 $R$이라면 $s$에선 $L$이 최대 $2c$번까지 등장할 수 있고 적어도 $c$번은 등장해야 한다. 확인한 $s$의 원소는 제거하자 (앞에서부터 확인하면 된다). 이를 만족하지 못하면 정답은 NO이다

`-` 만약 성공적으로 $p$의 순회가 끝났다면 $s$를 확인하자. $s$가 비어있으면 YES이고 그렇지 않으면 NO이다

```python
def solve_testcase(p, s):
    r_stack = l_stack = 0
    s = list(s[::-1])
    c = 0
    prev = p[0]
    for h in p:
        if h == prev:
            c += 2
            continue
        need = c // 2
        while s and s[-1] == prev:
            s.pop()
            c -= 1
            need -= 1
        if c < 0 or need > 0:
            return "NO"
        c = 2
        prev = h
    if s:
        return "NO"
    return "YES"


def solution():
    t = int(input())
    dummy = "#"
    for _ in range(t):
        p = input().rstrip()
        p += dummy
        s = input().rstrip()
        answer = solve_testcase(p, s)
        print(answer)


solution()

# input
# 2
# LR
# LLLR
# RRL
# RRRRL
```

## E. Boneca Ambalabu (Upsolving)

`-` 이 문제에 1시간은 매달렸지만 결국 풀지 못했다. 콘테스트가 끝나고 태그에 bitmasks만 있는 걸 보고 바로 풀 수 있었다 (그리디가 아니었어?)

`-` 임의의 $a_k$에 대해 $S_k(=(a_k \oplus a_1) + (a_k \oplus a_2) +\cdots + (a_k \oplus a_n))$가 어떻게 계산되는지 고찰해보자. $a_k$의 $i$번째 비트가 $x(=0 \text { or } 1)$일 때 $a_j$의 $i$번째 비트가 $x$라면 $a_k\oplus a_j$의 $i$번째 비트는 $0$이고 $1-x$라면 $1$이다. 그리고 이는 $2^i$의 크기를 가진다

`-` 만약 $i$번째 비트가 $1-x$인 $a_j$의 개수가 $c$라면 $S_k$에 $c2^i$만큼 기여한다. 여기서 중요한 것은 $i$번째 비트가 $1-x$인 $a_j$의 **개수**가 중요할 뿐이지 $a_j$가 무엇이냐는 중요하지 않다는 것이다. 이에 따라 처음 $1$번만 각 위치별로 해당 비트가 $1$인 원소의 개수를 계산해 놓자 ($0$인 비트 개수 $+$ $1$인 비트 개수 $= n$). 그렇게 하면 비트별로 $\oplus$ 연산은 독립이므로 $S_k$를 $O(\max a)$에 계산할 수 있다 

`-` 배열의 모든 원소는 $0$과 $2^{30}-1$ 사이의 값을 가진다. 따라서 $30$개의 비트만 관리하면 충분하므로 전체 알고리즘의 시간 복잡도는 $O(30n)$이다

```python
def solve_testcase(array):
    n = len(array)
    n_bits = 30
    bit_counts = [sum(1 for a in array if a & (1 << i)) for i in range(n_bits)]
    max_value = 0
    for a in array:
        value = 0
        for i in range(n_bits):
            if a & (1 << i):
                value += (n - bit_counts[i]) * (1 << i)
            else:
                value += bit_counts[i] * (1 << i)
        if max_value < value:
            max_value = value
    return max_value


def solution():
    t = int(input())
    for _ in range(t):
        n = int(input())
        a = list(map(int, input().split()))
        answer = solve_testcase(a)
        print(answer)


solution()

# input
# 2
# 5
# 8 13 4 5 15
# 6
# 625 676 729 784 841 900
```

`-` $\oplus$ 연산만 비트별로 독립인 게 아니라 $+$ 연산도 비트별로 독립인데 생각을 안 했다. 대충 생각하다 안 될거 같아서 때려치우고 그리디 알고리즘에 꽃혀서 시간을 버렸다

`-` 이게 $\oplus$ 연산은 해당 비트에만 영향을 끼치지만 덧셈은 더 큰 자릿수를 침범할 수 있어서 안 될것 같다고 생각했다. 예컨대 $70$과 $80$은 둘 다 십의 자리이지만 더하면 $150$으로 백의 자리에 영향을 끼친다. 이따구로 생각해서 그만뒀다

`-` 근데 결괏값에 끼치는 영향은 독립이니까 잘못된 생각이다. $170 + 180$은 $350$으로 백의 자리는 $200$만큼, 십의 자리는 $150$만큼 영향을 끼친다. 십의 자리가 $7,8$이 아니고 다른 거여도 백의 자리가 $200$만큼 영향을 끼치는 건 동일하다

`-` 뭔가 안 될것 같아 다른 접근을 시도할 땐 안 되는 근거를 정리해서 안 되는 이유를 명확하게 만들자

## G. Chimpanzini Bananini (01:45)

`-` E번 문제와 F번 문제를 붙잡고 삽질만 하다가 때려치우고 G번 문제를 풀었다. 실질적으론 G번 문제 해결에 20분 정도 걸렸다

`-` 배열 $a$의 합계를 $s$, *rizziness*를 $r$, 배열의 반전 여부를 $t(=0 \text{ or } 1)$로 나타내자

`-` 현재 배열 $a$의 길이를 $n$이라 하자. $3$번 연산은 $s = s + k, r = r + (n+1)k$를 수행하고 $k$를 배열 $a$의 맨 뒤에 삽입한다

`-` $1$번 연산은 $a_n$을 배열에서 꺼낸 뒤 이를 $a$ 앞에 삽입한다. 그럼 $r$은 *rizziness* 공식에서 $i$ 대신 $i+1$을 사용하고 $n$ 대신 $n-1$을 사용하여 도출한 결괏값에 $a_n$을 더한 것과 같다

`-` $a_n + \sum\limits_{i=1}^{n - 1} a_i \cdot (i + 1) = \sum\limits_{i=1}^{n - 1} a_i \cdot i + \sum\limits_{i=1}^{n} a_i = n\cdot a_n - \sum\limits_{i=1}^{n} a_i \cdot i + \sum\limits_{i=1}^{n} a_i$이므로 $1$번 연산을 거치면 $r = r - n\cdot a_n + s$가 된다

`-` $2$번 연산은 가우스 일화를 이용해 계산할 수 있다. 그럼 $r = (n+1)s - r$이 된다

`-` 문제는 $2$번 연산은 배열을 반전시킨다는 것이다. 이를 실제로 수행하면 최악의 경우 $O\left(q^2\right)$이므로 시간 초과이다. 대신 배열 $a$를 deque을 이용하여 나타낸 뒤 반전 여부만 추적하자

`-` 만약 $t=0$이면 그대로인 것이고 $t=1$이면 배열이 반전된 상태이므로 맨 뒤가 사실은 맨 앞이다. 따라서 pop과 append는 popleft와 appendleft로 대체하면 된다 

`-` 모든 연산은 $O(1)$이므로 전체 알고리즘의 시간 복잡도는 $O(q)$이다

```python
from collections import deque


def solve_testcase(queries):
    q = len(queries)
    array = deque([])
    s = r = t = 0
    answers = [0] * q
    for i, op in enumerate(queries):
        n = len(array)
        if len(op) == 2:
            k = int(op[1])
            if t == 0:
                array.append(k)
            else:
                array.appendleft(k)
            s += k
            r += (n + 1) * k
        elif op[0] == "1":
            if t == 0:
                a_n = array.pop()
                array.appendleft(a_n)
            else:
                a_n = array.popleft()
                array.append(a_n)
            r += -n * a_n + s
        else:
            r = s * (n + 1) - r
            t = 1 - t
        answers[i] = r
    return answers


def solution():
    t = int(input())
    for _ in range(t):
        q = int(input())
        queries = [input().rstrip().split() for _ in range(q)]
        answers = solve_testcase(queries)
        print("\n".join(map(str, answers)))


solution()

# input
# 1
# 5
# 3 1
# 3 2
# 1
# 3 4
# 2
```

## B. The 67th 6-7 Integer Problem (00:04)

`-` 가장 큰 값을 제외한 나머지에 $-1$을 곱한 뒤 합을 구하면 된다

```python
def solve_testcase(array):
    max_ = max(array)
    return max_ + -1 * (sum(array) - max_)


def solution():
    t = int(input())
    for _ in range(t):
        array = list(map(int, input().split()))
        answer = solve_testcase(array)
        print(answer)


solution()

# input
# 1
# 1 2 3 4 5 6 7
```

## C. The 67th Permutation Problem (00:10)

`-` $3n$개의 원소를 오름차순으로 정렬한 뒤 덱에 저장하자

`-` 덱에서 최솟값을 popeft하고 최댓값을 pop하자

`-` 그 후 최댓값을 다시 pop하면 이것이 중앙값이 된다

```python
from collections import deque


def solve_testcase(n):
    array = deque(range(1, 3 * n + 1))
    sum_ = 0
    p = []
    while len(array) >= 3:
        small = array.popleft()
        large = array.pop()
        median = array.pop()
        p.extend([small, median, large])
    return p


def solution():
    t = int(input())
    for _ in range(t):
        n = int(input())
        answer = solve_testcase(n)
        print(*answer)


solution()

# input
# 1
# 2
```

## D. The 67th OEIS Problem (00:30)

`-` 오름차순으로 정렬된 소수 배열 $p$를 고려하자

`-` 에라토스테네스의 체로 적당히 $20000$개의 소수 목록을 구하면 된다. 나는 이걸 예전에 구현해둔 코드를 복붙해서 사용했는데 이게 잘못된 구현이어서 $1$번 틀렸다...

`-` 아무튼 $a_i=p_i \cdot p_{i+1}$로 설정하면 가능한 모든 $i$에 대해 $\gcd(a_i, a_{i+1})$은 서로 다른 값을 가진다

`-` 쿼리가 들어올 때마다 전처리 해둔 $a$에서 슬라이싱하여 반환하면 된다

```python
def sieve_of_eratosthenes(n):
    m = int(n**0.5)
    is_primes = [False, False] + [True] * (n - 1)
    for i in range(2, m + 1):
        if not is_primes[i]:
            continue
        for j in range(i * i, n + 1, i):
            is_primes[j] = False
    prime_nums = [i for i, is_prime in enumerate(is_primes) if is_prime]
    return prime_nums


def make_permutation(n, prime_nums):
    i, j = 0, 1
    a = []
    while len(a) < n:
        p1 = prime_nums[i]
        p2 = prime_nums[j]
        a.append(p1 * p2)
        i += 1
        j += 1
    return a


def solution():
    t = int(input())
    prime_nums = sieve_of_eratosthenes(2 * 10**5)
    a = make_permutation(10000, prime_nums)
    for _ in range(t):
        n = int(input())
        answer = a[:n]
        print(*answer)


solution()

# input
# 1
# 4
```

## E. The 67th XOR Problem (Upsolving)

`-` 문제를 잘못 이해해서 못풀고 에디토리얼을 봤다. 풀만 했는데...

`-` 처음 배열은 $a_1,a_2,\dots, a_n$이다. $a_i$를 골라 $\oplus$ 연산을 적용해보자

`-` 그럼 $a_1 \oplus a_i, a_2 \oplus a_i, \dots , a_n \oplus a_i$이 된다

`-` 이후 임의의 원소를 골라 $\oplus$ 연산을 적용하면 임의의 $x$에 대해 $x \oplus x = 0$이므로 직전의 $ \oplus $ 연산을 적용한 $a_i$가 더 이상 배열에 영향을 끼치지 못한다

`-` 그럼 최종적으로 남은 원소 중 가장 큰 것은 $\max\limits_{i < j} \{a_i \oplus a_j\}$와 같고 이는 $O\left(n^2\right)$에 계산할 수 있다. 트라이를 사용하면 $O(n\log n)$에도 가능하다

```python
from itertools import combinations


def solve_testcase(array):
    max_xor = 0
    for a_i, a_j in combinations(array, 2):
        max_xor = max(a_i ^ a_j, max_xor)
    return max_xor


def solution():
    t = int(input())
    for _ in range(t):
        n = int(input())
        array = list(map(int, input().split()))
        answer = solve_testcase(array)
        print(answer)


solution()

# input
# 1
# 2
# 1 3
```

## F. The 67th Tree Problem (01:32)

`-` 임의의 트리에 대해 리프 노드는 $y$에 포함된다

`-` 전체 노드의 개수는 $x+y$이므로 $x$를 증가시킬려면 $y$를 감소시켜야 되고 이는 리프 노드의 수를 최소로 해야 하는 것과 같다

`-` 편향 트리를 생각해보자. 편향 트리의 리프 노드는 $1$개이다

`-` 이때 $x=y$ 또는 $x + 1 = y$가 된다

`-` 따라서 $x > y$라면 이를 만족하는 트리는 존재하지 않는다

`-` 그렇지 않더라도 $x=0$이고 $y$가 짝수라면 최소한 루트 노드는 $x$에 포함되므로 이를 만족하는 트리는 존재하지 않는다

`-` $y-x$가 짝수라면 $2x$개의 노드로 편향 트리를 만든 뒤 리프 노드에 $y-x$개의 리프 노드를 연결시키자

`-` 반대로 $y-x$가 홀수라면 $2x+1$개의 노드로 편향 트리를 만든 뒤 리프 노드에 $y-x-1$개의 리프 노드를 연결시키면 된다

```python
def solve_testcase(x, y):
    if x > y or (x == 0 and y % 2 == 0):
        return "NO", []
    r = x + x
    gap = y - x
    if (y - x) % 2 == 1:
        r += 1
        gap -= 1
    edges = []
    for i in range(1, r):
        edges.append((i, i + 1))
    for j in range(r + 1, r + 1 + gap):
        edges.append((r, j))
    return "YES", edges


def solution():
    t = int(input())
    for _ in range(t):
        x, y = map(int, input().split())
        yes_or_no, edges = solve_testcase(x, y)
        if yes_or_no == "NO":
            print("NO")
            continue
        print("YES")
        for u, v in edges:
            print(u, v)


solution()


# input
# 1
# 0 5
```
